```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_excel('flight_price.xlsx')
```

```python
df.head()
```

```python
df['Date_of_Jour'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month_of_Jour'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year_of_Jour'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.drop('Date_of_Journey',axis = 1 , inplace = True)
```

```python
df.head()
```

```python
df['Date_of_Jour'] = df['Date_of_Jour'].astype(int)
df['Month_of_Jour'] = df['Month_of_Jour'].astype(int)
df['Year_of_Jour'] = df['Year_of_Jour'].astype(int)
```

```python
df.info()
```

```python
df.head()
```

```python
df['Arrival_Time'] = df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df['Arrival_Hour'] = df['Arrival_Time'].str.split(':').str[0].astype(int)
df['Arrival_Min'] = df['Arrival_Time'].str.split(':').str[1].astype(int)
```

```python
df.drop('Arrival_Time',inplace = True,axis= 1)
```

```python
df.head()
```

```python
df.info()
```

```python
df['Dep_Hour'] = df['Dep_Time'].str.split(':').str[0].astype(int)
df['Dep_Min'] = df['Dep_Time'].str.split(':').str[1].astype(int)
```

```python
df.drop("Dep_Time",inplace = True , axis = 1)
```

```python
df.head(2)
```

```python
df['Dur_Hour'] = df['Duration'].str.extract(r"(\d+)h").fillna(0).astype(int)
df['Dur_Min'] = df['Duration'].str.extract(r"(\d+)m").fillna(0).astype(int)
```

```python
df.drop('Duration',axis = 1, inplace = True)
```

```python
df['Total_Duration_Min'] = df['Dur_Hour'] * 60 + df['Dur_Min']
```
