# Feature Scaling using StandardScaler


## What is happening in this cell?

### Training Data
```python
fit_transform()
```
- learns mean and standard deviation
- transforms training data

### Testing Data
```python
transform()
```
- uses SAME scaling learned from training data

## Why are we doing this?
To:
- standardize feature ranges
- improve model stability
- avoid data leakage


````python
# =============================================================================
# Why do we use fit_transform() on training data
# and only transform() on test data?
# =============================================================================

# fit()       -> Learns parameters from the data
# transform() -> Applies the learned transformation
# fit_transform() -> Performs both fit() + transform()

# Example:
# StandardScaler learns:
#   1. Mean
#   2. Standard Deviation
# from the TRAINING data only.

# -----------------------------------------------------------------------------#
# Correct Approach
# -----------------------------------------------------------------------------#

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Step 1:
# Learn scaling parameters from training data
# and transform the training data
X_train_scaled = scaler.fit_transform(X_train)

# Step 2:
# ONLY transform test data using the SAME parameters
# learned from training data
X_test_scaled = scaler.transform(X_test)

# -----------------------------------------------------------------------------#
# Why NOT fit_transform() on test data?
# -----------------------------------------------------------------------------#

# If we use fit_transform() on test data:
#
# X_test_scaled = scaler.fit_transform(X_test)
#
# then the scaler learns information from the test set too.
#
# This causes DATA LEAKAGE because:
# - Model indirectly gets knowledge about test data
# - Evaluation becomes unrealistic
# - Accuracy may appear falsely high

# -----------------------------------------------------------------------------#
# Important Rule
# -----------------------------------------------------------------------------#

# Training data -> fit_transform()
# Test data     -> transform()

# The model should learn ONLY from training data.
# Test data should remain completely unseen during training.

# -----------------------------------------------------------------------------#
# Formula used in StandardScaler
# -----------------------------------------------------------------------------#

# z = (x - mean) / standard_deviation

# where:
# mean and standard deviation are calculated ONLY from training data.
# =============================================================================
````

# Understanding R² Formula


## What is happening in this markdown?

R² formula is explained.

## Meaning of R²
R² measures:
- how much variance is explained by model

Range:
- 1 → perfect prediction
- 0 → poor prediction


## R square
Formula

**R^2 = 1 - SSR/SST**  

R^2 = coefficient of determination SSR = sum of squares of residuals SST = total sum of squares

SSR = sum of squares regression
SST = sum of squares Total
SST (Sum of Squares Total) measures the total variation in the target variable, while SSR (Sum of Squares Regression) measures the variation that your model successfully explains.

# Understanding Adjusted R²


## What is happening in this markdown?

Adjusted R² formula is explained.

## Why is this important?
Adjusted R² penalizes unnecessary features.

Useful mainly in:
- Multiple Linear Regression


**Adjusted R2 = 1 – [(1-R2)*(n-1)/(n-k-1)]**

where:

R2: The R2 of the model 

n: The number of observations 

k: The number of predictor variables
