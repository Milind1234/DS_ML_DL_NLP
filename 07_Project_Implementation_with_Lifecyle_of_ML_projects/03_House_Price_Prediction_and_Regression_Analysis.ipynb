# Binary Encoding

Machine learning algorithms require numerical input.

The binary categorical features are converted as follows:

- Yes → 1
- No → 0

This preserves the information while making the features suitable for regression models.

````python
yes_no_cols = ['mainroad', 'guestroom', 'basement',

               'hotwaterheating', 'airconditioning', 'prefarea']

for col in yes_no_cols:

    df[col] = df[col].map({'yes': 1, 'no': 0})
````

# One-Hot Encoding for Furnishing Status

The `furnishingstatus` feature contains three categories:

- Furnished
- Semi-Furnished
- Unfurnished

One-hot encoding converts these categories into numerical variables.

The parameter `drop_first=True` is used to avoid the Dummy Variable Trap and reduce multicollinearity.

````python
furnishing_dummies = pd.get_dummies(
    df['furnishingstatus'],
    drop_first=True,
    dtype=int
)

df = pd.concat([df, furnishing_dummies], axis=1)
df.drop('furnishingstatus', axis=1, inplace=True)
````

### Interpretation

The category "Furnished" becomes the reference category.

The generated variables indicate whether a property is:

- Semi-Furnished
- Unfurnished

If both variables are 0, the property belongs to the Furnished category.

# Outlier Quantification Using IQR

The Interquartile Range (IQR) method is used to quantify the number of observations lying outside the expected range.

This analysis helps determine whether outliers should be investigated further.

````python
num_cols = ['price', 'area', 'bedrooms', 'bathrooms', 'stories', 'parking']

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} outliers")
````

### Interpretation

Although several observations are identified as outliers, many correspond to realistic housing characteristics such as:

- Large property areas
- Multiple stories
- Higher parking capacity

Therefore, no outliers are removed at this stage.

# Feature Matrix and Target Variable

The dataset is divided into:

- X (Independent Variables)

````python
X=df.drop('price',axis = 1)
y= df['price']
````

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=42)
````

### Interpretation

In an earlier iteration of the model, the Bedrooms feature exhibited:

- High VIF (> 10)
- High p-value (> 0.05)

Therefore, it was removed to reduce multicollinearity and improve model stability.

````python
X_train = X_train.drop('bedrooms', axis=1)
X_test = X_test.drop('bedrooms', axis=1)
X_train = X_train.drop('semi-furnished', axis=1)
X_test = X_test.drop('semi-furnished', axis=1)
````

# Multiple Linear Regression Model

A Multiple Linear Regression model is developed to examine the relationship between house prices and property features.

The model estimates:

- Feature coefficients
- Statistical significance
- Overall explanatory power

````python
import statsmodels.api as sm

X_train_sm = sm.add_constant(X_train)

model = sm.OLS(y_train, X_train_sm).fit()

print(model.summary())
````

### Interpretation

The regression summary provides:

- R² and Adjusted R²
- Coefficients
- P-values
- F-statistic

Key findings:

- Area, bathrooms, stories, parking, air conditioning, and preferred area significantly influence house prices.
- Features with high p-values contribute little predictive value and may be removed during feature selection.

# Multicollinearity Analysis Using VIF

Variance Inflation Factor (VIF) measures how strongly a feature is correlated with other predictors.

General guidelines:

- VIF < 5 → Acceptable
- VIF 5–10 → Moderate concern
- VIF > 10 → High multicollinearity

````python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif = pd.DataFrame()

vif["Feature"] = X_train.columns

vif["VIF"] = [
    variance_inflation_factor(X_train.values, i)
    for i in range(X_train.shape[1])
]

vif
````

### Interpretation

The Bedrooms feature exhibited a very high VIF value and an insignificant p-value.

Therefore, it was removed from the model to reduce multicollinearity and improve model stability.

# Residual Diagnostics

Residuals are calculated as:

Residual = Actual Price − Predicted Price

Residual analysis is performed to verify whether the assumptions of linear regression are reasonably satisfied.

````python
y_train_pred = model.predict(X_train_sm)

residuals = y_train - y_train_pred
````

````python
plt.figure(figsize=(8,5))

sns.histplot(residuals, kde=True)

plt.title("Residual Distribution")
plt.xlabel("Residuals")

plt.show()
````

### Interpretation

The residual distribution is approximately centered around zero.

This indicates that the model does not systematically overestimate or underestimate house prices.

A slight positive skew is observed, suggesting the model occasionally underestimates premium properties.

````python
plt.figure(figsize=(8,5))

plt.scatter(y_train_pred, residuals)

plt.axhline(y=0, color='red')

plt.xlabel("Predicted Price")
plt.ylabel("Residuals")

plt.title("Residuals vs Predicted")

plt.show()
````

### Interpretation

The residuals appear randomly distributed around zero.

This indicates:

- The linearity assumption is reasonably satisfied.
- No major systematic pattern exists in the residuals.

A slight increase in variance for higher predicted prices suggests mild heteroscedasticity.

# Regularization Techniques

In this section, we apply MinMax Scaling and compare:
- Linear Regression
- Ridge Regression
- Lasso Regression
- Elastic Net Regression

The models will be evaluated using:
- R² Score
- MAE
- MSE
- RMSE

````python
X = df[['area','bathrooms','stories','mainroad',
        'guestroom','basement','hotwaterheating',
        'airconditioning','parking','prefarea',
        'unfurnished']]

y = df['price']
````

````python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=42)
````

# Feature Scaling Using MinMaxScaler

Regularization techniques are sensitive to feature magnitude.

MinMaxScaler transforms numerical features to a range between 0 and 1, ensuring that all predictors contribute equally during model training.

````python
from sklearn.preprocessing import MinMaxScaler

num_cols = ['area','bathrooms','stories','parking']

scaler = MinMaxScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[num_cols] = scaler.fit_transform(
    X_train[num_cols]
)

X_test_scaled[num_cols] = scaler.transform(
    X_test[num_cols]
)
````

# Regularization Techniques

Three regularized regression models are evaluated:

## Ridge Regression
Penalizes large coefficients using L2 regularization.

## Lasso Regression
Penalizes coefficients using L1 regularization and can perform feature selection.

## Elastic Net Regression
Combines L1 and L2 penalties, balancing Ridge and Lasso behavior.

These models are compared against standard Linear Regression.

````python
from sklearn.linear_model import (
    LinearRegression,
    Ridge,
    Lasso,
    ElasticNet
)
````

````python
models = {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=1000),
    "Elastic Net": ElasticNet(alpha=1000, l1_ratio=0.5)
}
````

````python
import pandas as pd
import numpy as np

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

results = []

for name, model in models.items():

    model.fit(X_train_scaled, y_train)

    y_pred = model.predict(X_test_scaled)

    r2 = r2_score(y_test, y_pred)

    mae = mean_absolute_error(y_test, y_pred)

    mse = mean_squared_error(y_test, y_pred)

    rmse = np.sqrt(mse)

    results.append([
        name,
        r2,
        mae,
        mse,
        rmse
    ])

results_df = pd.DataFrame(
    results,
    columns=[
        'Model',
        'R2 Score',
        'MAE',
        'MSE',
        'RMSE'
    ]
)

results_df
````

````python
results_df.sort_values(
    by='R2 Score',
    ascending=False
)
````

### Interpretation

The model comparison table evaluates:

- R² Score
- MAE
- MSE
- RMSE

The best-performing model is identified based on its ability to achieve:

- Higher R²
- Lower prediction error

Regularization may improve generalization by reducing overfitting and controlling coefficient magnitudes.

````python
from sklearn.model_selection import GridSearchCV

ridge = Ridge()

params = {
    'alpha': [0.01, 0.1, 1, 10, 100, 1000]
}

ridge_cv = GridSearchCV(
    ridge,
    params,
    cv=5,
    scoring='r2'
)

ridge_cv.fit(X_train_scaled, y_train)

print("Best Alpha:", ridge_cv.best_params_)
print("Best Score:", ridge_cv.best_score_)
````

````python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
import numpy as np

ridge = Ridge()

ridge_params = {
    'alpha': [0.01, 0.1, 1, 10, 100, 1000]
}

ridge_cv = GridSearchCV(
    estimator=ridge,
    param_grid=ridge_params,
    cv=5,
    scoring='r2'
)

ridge_cv.fit(X_train_scaled, y_train)

print("Best Alpha:", ridge_cv.best_params_)
print("Best CV Score:", ridge_cv.best_score_)

best_ridge = ridge_cv.best_estimator_

y_pred_ridge = best_ridge.predict(X_test_scaled)

print("\nRIDGE RESULTS")
print("R²:", r2_score(y_test, y_pred_ridge))
print("MAE:", mean_absolute_error(y_test, y_pred_ridge))
print("MSE:", mean_squared_error(y_test, y_pred_ridge))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_ridge)))
````

````python
from sklearn.linear_model import Lasso

lasso = Lasso(max_iter=10000)

lasso_params = {
    'alpha': [0.01, 0.1, 1, 10, 100, 1000]
}

lasso_cv = GridSearchCV(
    estimator=lasso,
    param_grid=lasso_params,
    cv=5,
    scoring='r2'
)

lasso_cv.fit(X_train_scaled, y_train)

print("Best Alpha:", lasso_cv.best_params_)
print("Best CV Score:", lasso_cv.best_score_)

best_lasso = lasso_cv.best_estimator_

y_pred_lasso = best_lasso.predict(X_test_scaled)

print("\nLASSO RESULTS")
print("R²:", r2_score(y_test, y_pred_lasso))
print("MAE:", mean_absolute_error(y_test, y_pred_lasso))
print("MSE:", mean_squared_error(y_test, y_pred_lasso))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_lasso)))
````

````python
from sklearn.linear_model import ElasticNet

elastic = ElasticNet(max_iter=10000)

elastic_params = {
    'alpha': [0.01, 0.1, 1, 10, 100],
    'l1_ratio': [0.1, 0.3, 0.5, 0.7, 0.9]
}

elastic_cv = GridSearchCV(
    estimator=elastic,
    param_grid=elastic_params,
    cv=5,
    scoring='r2'
)

elastic_cv.fit(X_train_scaled, y_train)

print("Best Parameters:", elastic_cv.best_params_)
print("Best CV Score:", elastic_cv.best_score_)

best_elastic = elastic_cv.best_estimator_

y_pred_elastic = best_elastic.predict(X_test_scaled)

print("\nELASTIC NET RESULTS")
print("R²:", r2_score(y_test, y_pred_elastic))
print("MAE:", mean_absolute_error(y_test, y_pred_elastic))
print("MSE:", mean_squared_error(y_test, y_pred_elastic))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_elastic)))
````

````python
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Ridge', 'Lasso', 'Elastic Net'],
    'R2': [
        r2_linear,
        r2_score(y_test, y_pred_ridge),
        r2_score(y_test, y_pred_lasso),
        r2_score(y_test, y_pred_elastic)
    ],
    'MAE': [
        mae_linear,
        mean_absolute_error(y_test, y_pred_ridge),
        mean_absolute_error(y_test, y_pred_lasso),
        mean_absolute_error(y_test, y_pred_elastic)
    ],
    'RMSE': [
        rmse_linear,
        np.sqrt(mean_squared_error(y_test, y_pred_ridge)),
        np.sqrt(mean_squared_error(y_test, y_pred_lasso)),
        np.sqrt(mean_squared_error(y_test, y_pred_elastic))
    ]
})

results.sort_values('R2', ascending=False)
````
