## Implementing Decision Tree Classifier

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
##independent features
import pandas as pd
X= pd.DataFrame(iris['data'],columns=['sepal length (cm)',
 'sepal width (cm)',
 'petal length (cm)',
 'petal width (cm)'])
```

```python
##dependent features
y= iris['target']
```

```python
[X]
```

```python
##train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
treeclassifier = DecisionTreeClassifier(random_state=42)
```

```python
treeclassifier.fit(X_train,y_train)
```

```python
##Visualize the Decision Tree
import matplotlib.pyplot as plt
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treeclassifier,filled=True)
```

```python
y_pred = treeclassifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
print("confusion_matrix: \n",confusion_matrix(y_test,y_pred))
print("classification_report: \n",classification_report(y_test,y_pred))
```

## Decision Tree Classifier Prepruning And Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'criterion': ['gini', 'entropy', 'log_loss'],
    'splitter': ['best', 'random'],
    'max_depth': [None, 2, 3, 4, 5, 6, 8, 10],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': [None, 'sqrt', 'log2']
}
grid = GridSearchCV(treeclassifier,param_grid,scoring='accuracy',cv=5,refit=True,verbose=3)
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred1 = grid.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
print('acc_score : ' , accuracy_score(y_test,y_pred1))
print('confusion_matrix:\n',confusion_matrix(y_test,y_pred1))
print("classification_report:\n",classification_report(y_test,y_pred1))
```
