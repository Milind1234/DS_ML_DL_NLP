🔤 NLTK Word Tokenizers

Tokenization is the process of breaking text into smaller units called tokens.
A token can be a word, punctuation mark, number, or part of a word.

⸻

📌 Tokenizers Covered

1. WordPunctTokenizer().tokenize(corpus)
2. word_tokenize(corpus)
3. wordpunct_tokenize(corpus)
4. TreebankWordTokenizer().tokenize(corpus)

To understand the difference, consider:

Input:

I'm learning NLP. Don't worry!

⸻

1️⃣ WordPunctTokenizer().tokenize(corpus)

💡 What does it do?

WordPunctTokenizer separates text into words and punctuation marks using a simple regular-expression-based approach.

🔍 Example

Input:

I'm learning NLP. Don't worry!

Output:

['I', "'", 'm', 'learning', 'NLP', '.', 'Don', "'", 't', 'worry', '!']

🔎 How does it handle contractions?

Input	Output
I'm	['I', "'", 'm']
Don't	['Don', "'", 't']
It's	['It', "'", 's']

✅ Key Points

* Separates punctuation aggressively.
* Every punctuation mark becomes a separate token.
* Does not preserve common English contraction structure.
* Does not require sentence tokenization first.
* Useful for simple word-and-punctuation splitting.

🎯 Best Use

Use it when you want simple tokenization where punctuation should be completely separated from words.

⸻

2️⃣ word_tokenize(corpus)

💡 What does it do?

word_tokenize() is NLTK’s commonly used general-purpose word tokenizer.

By default, it first performs sentence tokenization and then applies Treebank-style word tokenization.

🔍 Example

Input:

I'm learning NLP. Don't worry!

Output:

['I', "'m", 'learning', 'NLP', '.', 'Do', "n't", 'worry', '!']

🔎 How does it handle contractions?

Input	Output
I'm	['I', "'m"]
Don't	['Do', "n't"]
It's	['It', "'s"]

Instead of splitting the apostrophe separately:

Don't ❌ ['Don', "'", 't']

it uses Treebank-style rules:

Don't ✅ ['Do', "n't"]

✅ Key Points

* General-purpose NLTK tokenizer.
* Handles common English contractions using Treebank-style rules.
* Separates most punctuation marks.
* Sentence-aware by default.
* Usually the most convenient choice for basic NLP learning.

🎯 Best Use

Use it for general English NLP preprocessing when you want sensible handling of punctuation and contractions.

⸻

3️⃣ wordpunct_tokenize(corpus)

💡 What does it do?

wordpunct_tokenize() is a convenient function that provides the same tokenization behavior as WordPunctTokenizer.

🔍 Example

Input:

I'm learning NLP. Don't worry!

Output:

['I', "'", 'm', 'learning', 'NLP', '.', 'Don', "'", 't', 'worry', '!']

🔗 Important Relationship

wordpunct_tokenize(corpus) ≈ WordPunctTokenizer().tokenize(corpus)

Both produce the same style of tokenization.

🔎 Contraction Handling

Input	Output
I'm	['I', "'", 'm']
Don't	['Don', "'", 't']
It's	['It', "'", 's']

✅ Key Points

* Aggressively separates punctuation.
* Does not preserve common English contraction structure.
* Simpler function-based syntax.
* Essentially a shortcut for using WordPunctTokenizer.

🎯 Best Use

Use it when you want the behavior of WordPunctTokenizer with a shorter and simpler syntax.

⸻

4️⃣ TreebankWordTokenizer().tokenize(corpus)

💡 What does it do?

TreebankWordTokenizer tokenizes text according to Penn Treebank tokenization conventions.

It is particularly useful for handling English contractions and punctuation according to standardized rules.

🔍 Example

Input:

I'm learning NLP. Don't worry!

Output:

['I', "'m", 'learning', 'NLP.', 'Do', "n't", 'worry', '!']

🔎 Contraction Handling

Input	Output
I'm	['I', "'m"]
Don't	['Do', "n't"]
It's	['It', "'s"]

⚠️ Important Difference

Direct TreebankWordTokenizer usage does not perform sentence tokenization first.

Therefore, sentence-final periods can sometimes remain attached to the preceding word:

NLP. → ['NLP.']

whereas word_tokenize() typically gives:

NLP. → ['NLP', '.']

✅ Key Points

* Uses Penn Treebank tokenization conventions.
* Handles common English contractions well.
* Does not automatically perform sentence tokenization.
* Can treat sentence-final periods differently from word_tokenize().

🎯 Best Use

Use it when you specifically need Penn Treebank-style tokenization.

⸻

📊 Quick Comparison

Tokenizer	Contractions	Punctuation	Sentence-Aware	Main Use
WordPunctTokenizer()	❌ Aggressive split	Fully separated	❌ No	Simple tokenization
word_tokenize()	✅ Treebank-style	Well handled	✅ Yes	General NLP
wordpunct_tokenize()	❌ Aggressive split	Fully separated	❌ No	Quick simple tokenization
TreebankWordTokenizer()	✅ Treebank-style	Treebank rules	❌ No	Treebank tokenization

⸻

🧪 Side-by-Side Example

For the word:

Don't

Tokenizer	Result
WordPunctTokenizer()	['Don', "'", 't']
wordpunct_tokenize()	['Don', "'", 't']
word_tokenize()	['Do', "n't"]
TreebankWordTokenizer()	['Do', "n't"]

The biggest difference is therefore:

WordPunct style

Don't → Don + ' + t

Treebank style

Don't → Do + n't

⸻

🔗 Relationship Between the Tokenizers

WordPunctTokenizer Family

WordPunctTokenizer().tokenize(corpus)
⬇️ Same behavior
wordpunct_tokenize(corpus)

Both use the same word-and-punctuation splitting approach.

word_tokenize() and TreebankWordTokenizer

Conceptually:

word_tokenize()

⬇️

Sentence Tokenization

⬇️

Treebank-Style Word Tokenization

This is why word_tokenize() and TreebankWordTokenizer() handle contractions similarly but may differ in how they handle sentence boundaries.

🏆 Recommendation

Requirement	Recommended Tokenizer
General NLP learning	⭐ word_tokenize()
Simple punctuation splitting	wordpunct_tokenize()
Object-oriented tokenizer usage	WordPunctTokenizer()
Penn Treebank-style tokenization	TreebankWordTokenizer()

For most basic NLTK NLP projects, word_tokenize() is a convenient default.
The other tokenizers are useful when you need specific tokenization behavior.

```python
##  Tokenization
## Paragraphs ---> Sentence
from nltk.tokenize import sent_tokenize
```

```python
corpus="""The ancient lighthouse stood as a silent sentinel against the crashing waves, its whitewashed stone walls weathered by centuries of salty sea breeze. Inside, a spiral staircase wound its way toward the lantern room, where a meticulously polished brass mechanism awaited the evening light. Outside, the sky transformed from a vibrant amber to a deep twilight indigo, painting the horizon in quiet, sweeping strokes of color."""
```

```python
print(corpus)
```

```python
##  Tokenization
## Paragraphs ---> Sentence
document = sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
    print(sentence)
```

```python
## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for words in document:
    print(word_tokenize(words))
```

```python
from nltk.tokenize import WordPunctTokenizer
```

```python
WordPunctTokenizer().tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
TreebankWordTokenizer().tokenize(corpus)
```
