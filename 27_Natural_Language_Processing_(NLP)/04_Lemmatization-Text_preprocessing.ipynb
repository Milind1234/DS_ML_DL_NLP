## Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −

### One important point: WordNetLemmatizer directly supports only these WordNet POS categories:

* n → Noun
* v → Verb
* a → Adjective
* r → Adverb

It does not have dedicated POS parameters for pronouns, prepositions, conjunctions, etc.

## 1. Clear practical example with different POS types

```python
from nltk.stem import WordNetLemmatizer

# Creating the WordNet Lemmatizer object
lemmatizer = WordNetLemmatizer()


# Different words with their appropriate WordNet POS tags
words_with_pos = [
    
    # -------------------------
    # NOUNS -> POS = "n"
    # -------------------------
    ("cars", "n"),
    ("children", "n"),
    ("mice", "n"),
    ("geese", "n"),
    ("feet", "n"),
    ("women", "n"),

    # -------------------------
    # VERBS -> POS = "v"
    # -------------------------
    ("running", "v"),
    ("ran", "v"),
    ("went", "v"),
    ("eating", "v"),
    ("written", "v"),
    ("studying", "v"),

    # -------------------------
    # ADJECTIVES -> POS = "a"
    # -------------------------
    ("better", "a"),
    ("best", "a"),
    ("worse", "a"),
    ("happier", "a"),
    ("stronger", "a"),

    # -------------------------
    # ADVERBS -> POS = "r"
    # -------------------------
    ("quickly", "r"),
    ("slowly", "r"),
    ("carefully", "r"),
    ("happily", "r"),
]


# Lemmatize each word according to its POS
for word, pos in words_with_pos:
    lemma = lemmatizer.lemmatize(word, pos=pos)
    print(f"{word:<12} POS={pos} ---> {lemma}")
```

Notice that lemmatization does not guarantee every word will change. For example, many adverbs such as quickly remain unchanged because WordNet does not necessarily map them to an adjective like quick.

## 2. Compare the same word using different POS parameters

This is an excellent example for understanding why the correct POS matters.

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "running",
    "better",
    "studies",
    "playing",
    "cars"
]

for word in words:
    print(f"\nWord: {word}")
    print("-" * 30)

    print("Noun      :", lemmatizer.lemmatize(word, pos="n"))
    print("Verb      :", lemmatizer.lemmatize(word, pos="v"))
    print("Adjective :", lemmatizer.lemmatize(word, pos="a"))
    print("Adverb    :", lemmatizer.lemmatize(word, pos="r"))
```

## 3. Better structured implementation using WordNet constants

Instead of directly writing "n", "v", "a", and "r", you can use WordNet constants:

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

lemmatizer = WordNetLemmatizer()

words_with_pos = [
    # Nouns
    ("cars", wordnet.NOUN),
    ("children", wordnet.NOUN),
    ("mice", wordnet.NOUN),

    # Verbs
    ("running", wordnet.VERB),
    ("went", wordnet.VERB),
    ("written", wordnet.VERB),

    # Adjectives
    ("better", wordnet.ADJ),
    ("worse", wordnet.ADJ),
    ("stronger", wordnet.ADJ),

    # Adverbs
    ("quickly", wordnet.ADV),
    ("slowly", wordnet.ADV),
    ("carefully", wordnet.ADV)
]

for word, pos in words_with_pos:
    lemma = lemmatizer.lemmatize(word, pos=pos)

    print(
        f"Original: {word:<12} "
        f"POS: {pos:<3} "
        f"Lemma: {lemma}"
    )
```

## 4. Practical use case: sentence lemmatization with automatic POS detection

In real NLP applications, you usually don’t manually specify the POS of every word.

Example:-
        The children were running quickly and eating better foods.

A practical pipeline is:

                                    Raw Sentence
                                         ↓
                                    Tokenization
                                         ↓
                                    POS Tagging
                                         ↓
                                    Convert NLTK POS → WordNet POS
                                         ↓
                                    Lemmatization
                                         ↓
                                    Lemmatized Sentence

```python
import nltk

# Required for word_tokenize()
nltk.download("punkt")
nltk.download("punkt_tab")

# Required for pos_tag()
nltk.download("averaged_perceptron_tagger_eng")

# Required for WordNetLemmatizer
nltk.download("wordnet")

# Optional: additional WordNet multilingual data
nltk.download("omw-1.4")
```

```python
import nltk

from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet


# Create lemmatizer
lemmatizer = WordNetLemmatizer()


# Function to convert NLTK POS tags
# into WordNet POS tags
def get_wordnet_pos(tag):

    if tag.startswith("J"):
        return wordnet.ADJ

    elif tag.startswith("V"):
        return wordnet.VERB

    elif tag.startswith("N"):
        return wordnet.NOUN

    elif tag.startswith("R"):
        return wordnet.ADV

    else:
        # Default POS for WordNetLemmatizer
        return wordnet.NOUN


# Example sentence
sentence = "The children were running quickly and eating better foods."


# Step 1: Tokenization
tokens = word_tokenize(sentence)


# Step 2: POS Tagging
pos_tags = pos_tag(tokens)


# Step 3: Lemmatization using the detected POS
lemmatized_words = []

for word, tag in pos_tags:

    wordnet_pos = get_wordnet_pos(tag)

    # Lemmatize only if WordNet supports the POS
    if wordnet_pos:
        lemma = lemmatizer.lemmatize(
            word.lower(),
            pos=wordnet_pos
        )
    else:
        # Keep unsupported POS unchanged
        lemma = word.lower()

    lemmatized_words.append(lemma)


# Display results
print("Original Tokens:")
print(tokens)

print("\nPOS Tags:")
print(pos_tags)

print("\nLemmatized Words:")
print(lemmatized_words)

print("\nLemmatized Sentence:")
print(" ".join(lemmatized_words))
```

# 🧠 POS-Aware Lemmatization Using `WordNetLemmatizer`

## 📌 What is Lemmatization?

**Lemmatization** is the process of reducing an inflected word to its meaningful **dictionary or base form**, known as a **lemma**.

### Examples

| Original Word | Lemma |
|---|---|
| `children` | `child` |
| `were` | `be` |
| `running` | `run` |
| `eating` | `eat` |
| `better` | `good` |
| `foods` | `food` |

> **Lemmatization vs Stemming:** Stemming simply removes prefixes or suffixes using rules, whereas lemmatization uses vocabulary and linguistic information to return meaningful dictionary words.

---

# 🔄 Complete NLP Pipeline

The POS-aware lemmatization process follows these steps:

**Raw Sentence**  
⬇️  
**1. Tokenization** — Split the sentence into tokens  
⬇️  
**2. POS Tagging** — Identify the grammatical role of each token  
⬇️  
**3. POS Mapping** — Convert NLTK POS tags into WordNet POS tags  
⬇️  
**4. Lemmatization** — Convert words into their base forms  
⬇️  
**Preprocessed Text**

---

# 📝 Example Sentence

`The children were running quickly and eating better foods.`

The objective is to reduce inflected words such as `children`, `were`, `running`, and `better` to their correct base forms.

---

# 1️⃣ Step 1: Tokenization

Tokenization splits the sentence into individual tokens.

`word_tokenize(sentence)`

### Output

`['The', 'children', 'were', 'running', 'quickly', 'and', 'eating', 'better', 'foods', '.']`

### What Happened?

| Original Text | Tokenized |
|---|---|
| `The children...` | `The`, `children`, `were`, `running`, ... |
| `foods.` | `foods`, `.` |

The punctuation mark `.` is also treated as a separate token.

---

# 2️⃣ Step 2: POS Tagging

POS stands for **Part of Speech**.

`pos_tag()` analyzes each token and identifies its grammatical role.

### Output

`[('The', 'DT'), ('children', 'NNS'), ('were', 'VBD'), ('running', 'VBG'), ('quickly', 'RB'), ('and', 'CC'), ('eating', 'VBG'), ('better', 'JJR'), ('foods', 'NNS'), ('.', '.')]`

### Understanding the POS Tags

| Word | POS Tag | Meaning |
|---|---|---|
| `The` | `DT` | Determiner |
| `children` | `NNS` | Plural Noun |
| `were` | `VBD` | Past-Tense Verb |
| `running` | `VBG` | Gerund / Present Participle Verb |
| `quickly` | `RB` | Adverb |
| `and` | `CC` | Coordinating Conjunction |
| `eating` | `VBG` | Gerund / Present Participle Verb |
| `better` | `JJR` | Comparative Adjective |
| `foods` | `NNS` | Plural Noun |
| `.` | `.` | Punctuation |

> ⚠️ **Important:** `pos_tag()` returns **Penn Treebank POS tags**, but `WordNetLemmatizer` expects **WordNet POS tags**. Therefore, we need to convert them.

---

# 3️⃣ Step 3: POS Tag Conversion

`WordNetLemmatizer` mainly supports four Part-of-Speech categories:

| NLTK POS Starts With | Category | WordNet POS |
|---|---|---|
| `J` | Adjective | `wordnet.ADJ` |
| `V` | Verb | `wordnet.VERB` |
| `N` | Noun | `wordnet.NOUN` |
| `R` | Adverb | `wordnet.ADV` |

### Conversion Examples

`NNS → N → wordnet.NOUN`

`VBG → V → wordnet.VERB`

`JJR → J → wordnet.ADJ`

`RB → R → wordnet.ADV`

Therefore:

| Word | NLTK POS | Converted WordNet POS |
|---|---|---|
| `children` | `NNS` | `NOUN` |
| `were` | `VBD` | `VERB` |
| `running` | `VBG` | `VERB` |
| `quickly` | `RB` | `ADVERB` |
| `better` | `JJR` | `ADJECTIVE` |
| `foods` | `NNS` | `NOUN` |

### What About Unsupported POS Tags?

Some POS categories do not have a corresponding WordNet lemmatizer category.

For example:

| Word | POS | Category | Action |
|---|---|---|---|
| `The` | `DT` | Determiner | Keep unchanged |
| `and` | `CC` | Conjunction | Keep unchanged |
| `.` | `.` | Punctuation | Keep unchanged |

---

# 4️⃣ Step 4: POS-Aware Lemmatization

After determining the correct WordNet POS, each supported word is lemmatized.

### 🔹 Noun Lemmatization

`children → NNS → NOUN → child`

`foods → NNS → NOUN → food`

The plural nouns are converted into their singular dictionary forms.

---

### 🔹 Verb Lemmatization

`were → VBD → VERB → be`

`running → VBG → VERB → run`

`eating → VBG → VERB → eat`

The different verb forms are converted into their infinitive/base forms.

---

### 🔹 Adjective Lemmatization

`better → JJR → ADJECTIVE → good`

This is an important example because `better` is an irregular comparative adjective.

With the correct POS:

`better + ADJECTIVE → good`

Without the correct POS:

`better → better`

This demonstrates why providing the correct POS is important for accurate lemmatization.

---

### 🔹 Adverb Lemmatization

`quickly → RB → ADVERB → quickly`

The word remains unchanged because `quickly` is already treated as its lemma.

> Lemmatization does **not** necessarily change every word.

---

### 🔹 Unsupported POS

`The → DT → Unsupported → the`

`and → CC → Unsupported → and`

`. → Punctuation → Unsupported → .`

These tokens are not processed by the WordNet lemmatizer and are kept unchanged. In this implementation, all words are also converted to lowercase.

---

| Original | NLTK POS | WordNet POS | Lemma |
|---|---|---|---|
| `The` | `DT` | — | `the` |
| `children` | `NNS` | Noun | `child` |
| `were` | `VBD` | Verb | `be` |
| `running` | `VBG` | Verb | `run` |
| `quickly` | `RB` | Adverb | `quickly` |
| `and` | `CC` | — | `and` |
| `eating` | `VBG` | Verb | `eat` |

| `foods` | `NNS` | Noun | `food` |
| `.` | `.` | — | `.` |

`The children were running quickly and eating better foods.`

### Tokenized Sentence

`['The', 'children', 'were', 'running', 'quickly', 'and', 'eating', 'better', 'foods', '.']`

### POS-Tagged Sentence

`[('The', 'DT'), ('children', 'NNS'), ('were', 'VBD'), ('running', 'VBG'), ('quickly', 'RB'), ('and', 'CC'), ('eating', 'VBG'), ('better', 'JJR'), ('foods', 'NNS'), ('.', '.')]`

# 🔑 Why Does POS Matter?

Consider the following words:

| Word | Default Lemmatization | With Correct POS |
|---|---|---|
| `running` | `running` | `run` |
| `were` | `were` | `be` |
| `better` | `better` | `good` |

By default, `WordNetLemmatizer` assumes that a word is a **noun**.

Therefore:

`lemmatize("running") → "running"`

But:

`lemmatize("running", pos=wordnet.VERB) → "run"`

This is why **POS-aware lemmatization is more accurate** than simply applying the lemmatizer directly to every word.

# 🧩 How the Code Works

The most important logic is:

`NLTK POS Tag → Check First Character → Convert to WordNet POS`

### Mapping Logic

`J... → Adjective`

`V... → Verb`

`N... → Noun`

`R... → Adverb`

For example:

`VBG → starts with V → wordnet.VERB`

`NNS → starts with N → wordnet.NOUN`

`JJR → starts with J → wordnet.ADJ`

`RB → starts with R → wordnet.ADV`

# 💡 Key Takeaways

- **Tokenization** breaks a sentence into individual tokens.
- **POS tagging** identifies the grammatical role of each token.
- NLTK's `pos_tag()` returns **Penn Treebank POS tags**.
- `WordNetLemmatizer` requires **WordNet-compatible POS tags**.
- The POS mapping function converts NLTK tags into WordNet tags.
- WordNet mainly supports **Nouns, Verbs, Adjectives, and Adverbs** for lemmatization.
- Unsupported POS categories can be kept unchanged.
- Providing the correct POS produces more accurate lemmas.
- The final lemmatized sentence does not need to be grammatically correct.
