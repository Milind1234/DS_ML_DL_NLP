# 🧹 Text Preprocessing: Stopword Removal with NLTK

## Objective

In this notebook, we will learn how to preprocess text using NLTK by applying:

1. **Sentence Tokenization**
2. **Word Tokenization**
3. **Lowercasing**
4. **Stopword Removal**
5. **Porter Stemming**
6. **Snowball Stemming**
7. **WordNet Lemmatization**

Stopwords are frequently occurring words such as `the`, `is`, `and`, and `of` that often provide little useful information for traditional NLP tasks.

> The original text will be preserved so that each preprocessing technique can be applied independently and compared fairly.

```python
## Speech Of DR APJ Abdul Kalam
paragraph = """I have three visions for India. In 3000 years of our history, people from all over 
               the world have come and invaded us, captured our lands, conquered our minds. 
               From Alexander onwards, the Greeks, the Turks, the Moguls, the Portuguese, the British,
               the French, the Dutch, all of them came and looted us, took over what was ours. 
               Yet we have not done this to any other nation. We have not conquered anyone. 
               We have not grabbed their land, their culture, 
               their history and tried to enforce our way of life on them. 
               Why? Because we respect the freedom of others.That is why my 
               first vision is that of freedom. I believe that India got its first vision of 
               this in 1857, when we started the War of Independence. It is this freedom that
               we must protect and nurture and build on. If we are not free, no one will respect us.
               My second vision for India’s development. For fifty years we have been a developing nation.
               It is time we see ourselves as a developed nation. We are among the top 5 nations of the world
               in terms of GDP. We have a 10 percent growth rate in most areas. Our poverty levels are falling.
               Our achievements are being globally recognised today. Yet we lack the self-confidence to
               see ourselves as a developed nation, self-reliant and self-assured. Isn’t this incorrect?
               I have a third vision. India must stand up to the world. Because I believe that unless India 
               stands up to the world, no one will respect us. Only strength respects strength. We must be 
               strong not only as a military power but also as an economic power. Both must go hand-in-hand. 
               My good fortune was to have worked with three great minds. Dr. Vikram Sarabhai of the Dept. of 
               space, Professor Satish Dhawan, who succeeded him and Dr. Brahm Prakash, father of nuclear material.
               I was lucky to have worked with all three of them closely and consider this the great opportunity of my life. 
               I see four milestones in my career"""
```

```python
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer

# Download required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
```

## 1. Initialize NLP Tools

We initialize the stopword collection and the different word normalization techniques once and reuse them throughout the notebook.

Using a `set` for stopwords makes repeated membership checks efficient.

```python
# Store English stopwords in a set for efficient lookup
english_stopwords = set(stopwords.words("english"))

# Initialize stemmers
porter_stemmer = PorterStemmer()
snowball_stemmer = SnowballStemmer("english")

# Initialize lemmatizer
lemmatizer = WordNetLemmatizer()

print(f"Number of English stopwords: {len(english_stopwords)}")
```

## 2. Sentence Tokenization

The paragraph is first divided into individual sentences.

We keep the original `sentences` unchanged and create separate processed outputs for each technique.

```python
# Split the original paragraph into sentences
sentences = sent_tokenize(paragraph)

print(f"Total sentences: {len(sentences)}")

# Display the first few sentences
for sentence in sentences[:3]:
    print(sentence)
    print()
```

## 3. Porter Stemming with Stopword Removal

The following steps are applied to each sentence:

**Sentence → Tokenization → Lowercasing → Stopword Removal → Porter Stemming → Rejoin Tokens**

`PorterStemmer` reduces words to their stems using rule-based suffix removal. The resulting stem may not always be a valid dictionary word.

```python
porter_sentences = []

for sentence in sentences:

    # Convert sentence into individual word tokens
    words = word_tokenize(sentence)

    # Remove stopwords and apply Porter stemming
    processed_words = [
        porter_stemmer.stem(word.lower())
        for word in words
        if word.lower() not in english_stopwords
    ]

    # Convert the processed tokens back into a sentence
    processed_sentence = " ".join(processed_words)

    porter_sentences.append(processed_sentence)
```

```python
for sentence in porter_sentences:
    print(sentence)
    print()
```

## 4. Snowball Stemming with Stopword Removal

`SnowballStemmer` is an improved and more systematic stemming algorithm based on the original Porter stemming algorithm.

The preprocessing pipeline is:

**Sentence → Tokenization → Lowercasing → Stopword Removal → Snowball Stemming → Rejoin Tokens**

```python
snowball_sentences = []

for sentence in sentences:

    # Tokenize the sentence
    words = word_tokenize(sentence)

    # Remove stopwords and apply Snowball stemming
    processed_words = [
        snowball_stemmer.stem(word.lower())
        for word in words
        if word.lower() not in english_stopwords
    ]

    # Reconstruct the processed sentence
    processed_sentence = " ".join(processed_words)

    snowball_sentences.append(processed_sentence)
```

```python
for sentence in snowball_sentences:
    print(sentence)
    print()
```

## 5. WordNet Lemmatization with Stopword Removal

Lemmatization reduces words to meaningful dictionary forms called **lemmas**.

Unlike stemming, it uses lexical information rather than simply removing suffixes.

In this basic example, words are lemmatized as verbs using `pos="v"`.

> For more accurate real-world preprocessing, POS-aware lemmatization should be used because not every word in a sentence is a verb.

```python
lemmatized_sentences = []

for sentence in sentences:

    # Tokenize the original sentence
    words = word_tokenize(sentence)

    # Remove stopwords and apply verb-based lemmatization
    processed_words = [
        lemmatizer.lemmatize(word.lower(), pos="v")
        for word in words
        if word.lower() not in english_stopwords
    ]

    # Reconstruct the processed sentence
    processed_sentence = " ".join(processed_words)

    lemmatized_sentences.append(processed_sentence)
```

```python
for sentence in lemmatized_sentences:
    print(sentence)
    print()
```

## 6. Comparing the Results

Each preprocessing technique was applied independently to the same original text.

| Technique | Purpose | Example |
|---|---|---|
| Stopword Removal | Removes common low-information words | `the`, `is`, `and` → removed |
| Porter Stemmer | Aggressively reduces words using rules | `studies` → `studi` |
| Snowball Stemmer | More refined stemming | `studies` → `studi` |
| WordNet Lemmatizer | Finds meaningful base forms | `studies` → `study` |

### Important Observation

The original `sentences` variable should **not be modified repeatedly**.

Incorrect workflow:

`Original → Porter Stemmed → Snowball Stemmed → Lemmatized`

Correct workflow:

`Original → Porter Stemmed`

`Original → Snowball Stemmed`

`Original → Lemmatized`

This ensures that every preprocessing technique is tested independently on the same original data.
