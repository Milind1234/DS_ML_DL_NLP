# 🏷️ Part-of-Speech (POS) Tagging with NLTK

## 📌 What is Part-of-Speech Tagging?

**Part-of-Speech (POS) Tagging** is the process of assigning a grammatical category to each word in a sentence based on its meaning and context.

Common grammatical categories include:

- **Noun** — person, place, object, or idea
- **Verb** — action or state
- **Adjective** — describes a noun
- **Adverb** — describes a verb, adjective, or another adverb
- **Pronoun** — replaces a noun
- **Determiner** — introduces or identifies a noun
- **Preposition** — shows relationships between words
- **Conjunction** — connects words or clauses

### Example

**Sentence:**

`Taj Mahal is a beautiful monument.`

After POS tagging, each word receives a grammatical tag:

`Taj → NNP`

`Mahal → NNP`

`is → VBZ`

`a → DT`

`beautiful → JJ`

`monument → NN`

# 1️⃣ Basic POS Tagging

Let's start with a simple sentence:

`Taj Mahal is a beautiful monument.`

The process is:

**Sentence → Word Tokenization → POS Tagging**

We use `word_tokenize()` instead of Python's `.split()` because NLTK tokenization handles punctuation and linguistic boundaries more appropriately.

```python
# Example sentence
sentence = "Taj Mahal is a beautiful monument."


# Step 1: Tokenize the sentence
tokens = word_tokenize(sentence)


# Step 2: Assign POS tags to each token
tagged_words = pos_tag(tokens)


# Display results
print("Original Sentence:")
print(sentence)

print("\nTokens:")
print(tokens)

print("\nPOS Tags:")
print(tagged_words)
```

## 🔍 Understanding the Output

| Word | POS Tag | Meaning |
|---|---|---|
| `Taj` | `NNP` | Proper Noun |
| `Mahal` | `NNP` | Proper Noun |
| `is` | `VBZ` | Third-Person Singular Present Verb |
| `a` | `DT` | Determiner |
| `beautiful` | `JJ` | Adjective |
| `monument` | `NN` | Singular Noun |
| `.` | `.` | Punctuation |

### Interpretation

`Taj Mahal` → **Proper Noun**

`is` → **Verb**

`a` → **Determiner**

`beautiful` → **Adjective**

`monument` → **Noun**

This demonstrates how POS tagging identifies the grammatical role of each word automatically.

```python
## Speech Of DR APJ Abdul Kalam
paragraph = """I have three visions for India. In 3000 years of our history, people from all over 
               the world have come and invaded us, captured our lands, conquered our minds. 
               From Alexander onwards, the Greeks, the Turks, the Moguls, the Portuguese, the British,
               the French, the Dutch, all of them came and looted us, took over what was ours. 
               Yet we have not done this to any other nation. We have not conquered anyone. 
               We have not grabbed their land, their culture, 
               their history and tried to enforce our way of life on them. 
               Why? Because we respect the freedom of others.That is why my 
               first vision is that of freedom. I believe that India got its first vision of 
               this in 1857, when we started the War of Independence. It is this freedom that
               we must protect and nurture and build on. If we are not free, no one will respect us.
               My second vision for India’s development. For fifty years we have been a developing nation.
               It is time we see ourselves as a developed nation. We are among the top 5 nations of the world
               in terms of GDP. We have a 10 percent growth rate in most areas. Our poverty levels are falling.
               Our achievements are being globally recognised today. Yet we lack the self-confidence to
               see ourselves as a developed nation, self-reliant and self-assured. Isn’t this incorrect?
               I have a third vision. India must stand up to the world. Because I believe that unless India 
               stands up to the world, no one will respect us. Only strength respects strength. We must be 
               strong not only as a military power but also as an economic power. Both must go hand-in-hand. 
               My good fortune was to have worked with three great minds. Dr. Vikram Sarabhai of the Dept. of 
               space, Professor Satish Dhawan, who succeeded him and Dr. Brahm Prakash, father of nuclear material.
               I was lucky to have worked with all three of them closely and consider this the great opportunity of my life. 
               I see four milestones in my career"""
```

```python
# Split the paragraph into individual sentences
sentences = sent_tokenize(paragraph)

# Store English stopwords in a set for efficient lookup
english_stopwords = set(stopwords.words("english"))

print(f"Total number of sentences: {len(sentences)}")
```

```python
# Store the POS-tagged results
tagged_sentences = []


for sentence in sentences:

    # Step 1: Tokenize the sentence
    words = word_tokenize(sentence)

    # Step 2: Remove English stopwords
    filtered_words = [
        word
        for word in words
        if word.lower() not in english_stopwords
    ]

    # Step 3: Perform POS tagging
    tagged_words = pos_tag(filtered_words)

    # Store the result
    tagged_sentences.append(tagged_words)
```

```python
# Display POS tags sentence by sentence
for index, tagged_words in enumerate(tagged_sentences, start=1):

    print(f"\nSentence {index}")
    print("-" * 60)

    for word, tag in tagged_words:
        print(f"{word:<20} ---> {tag}")
```

# ⚠️ Should We Remove Stopwords Before POS Tagging?

Not always.

Consider:

`The children are running in the park.`

Words such as `the`, `are`, and `in` may be considered stopwords, but they still contain useful grammatical information.

For example:

| Word | POS |
|---|---|
| `The` | Determiner |
| `are` | Verb |
| `in` | Preposition |

Therefore:

### For grammatical analysis

**Keep stopwords.**

`Sentence → Tokenization → POS Tagging`

### For some traditional NLP preprocessing tasks

Stopwords may be removed after POS tagging or according to the specific requirements of the task.

`Sentence → Tokenization → POS Tagging → Select Important Words`

> **Best Practice:** Perform POS tagging on the original tokenized sentence when grammatical context is important.

```python
# POS tagging while preserving the complete sentence structure
complete_pos_tags = []


for sentence in sentences:

    # Tokenize the complete sentence
    words = word_tokenize(sentence)

    # Perform POS tagging without removing stopwords
    tagged_words = pos_tag(words)

    # Store results
    complete_pos_tags.append(tagged_words)
```

```python
for index, tagged_words in enumerate(complete_pos_tags[:3], start=1):

    print(f"\nSentence {index}")
    print("-" * 60)

    for word, tag in tagged_words:
        print(f"{word:<20} ---> {tag}")
```

# 🧠 Key Takeaways

- **POS Tagging** identifies the grammatical role of each word.
- `word_tokenize()` converts a sentence into individual tokens.
- `pos_tag()` assigns Penn Treebank POS tags to those tokens.
- `NN` represents nouns, `VB` verbs, `JJ` adjectives, and `RB` adverbs.
- More specific tags such as `NNS`, `VBG`, and `JJR` provide additional grammatical information.
- Stopword removal is **not required** before POS tagging.
- Keeping stopwords usually preserves more grammatical context.
- For larger text, process each sentence independently for cleaner analysis.
