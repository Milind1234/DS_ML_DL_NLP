# 🧠 Word2Vec — Practical Implementation with Gensim

## 📌 What is Word2Vec?

**Word2Vec** is a word embedding technique that represents words as dense numerical vectors in a continuous vector space.

Unlike **Bag of Words (BoW)** and **TF-IDF**, where words are represented mainly based on frequency or importance, Word2Vec learns representations based on the **context in which words appear**.

The main idea is:

> Words that appear in similar contexts tend to have similar vector representations.

For example, words such as:

`king`, `queen`, `prince`, `royal`

may be positioned relatively close to each other in the embedding space because they occur in similar linguistic contexts.

---

# 🔢 From Words to Dense Vectors

A word such as:

`king`

can be represented as a numerical vector:

`king → [0.12, -0.34, 0.78, ..., 0.41]`

In the pretrained model used in this notebook, every known word is represented using a **300-dimensional vector**.

Therefore:

`king → 300 numerical values`

These numerical vectors allow us to perform mathematical operations on words and measure semantic relationships.

# 📦 1. Import Required Libraries

We use **Gensim**, a Python library designed for topic modeling and vector-space representations of text.

`gensim.downloader` provides access to several pretrained word embedding models.

In this notebook, we use:

`word2vec-google-news-300`

This is a pretrained Word2Vec model containing **300-dimensional word embeddings**.

```python
import gensim.downloader as api
```

# 📥 2. Load the Pretrained Word2Vec Model

Instead of training Word2Vec from scratch, we load a pretrained model.

The model used here is:

`word2vec-google-news-300`

The number `300` indicates that every word is represented using a vector containing **300 numerical dimensions**.

### Why Use a Pretrained Model?

Training high-quality word embeddings requires:

- A very large text corpus
- Significant computational resources
- Considerable training time

A pretrained model allows us to directly explore learned semantic relationships between words.

> ⚠️ The model is large and may require significant memory and download time the first time it is loaded.

```python
# Load the pretrained Google News Word2Vec model
wv = api.load("word2vec-google-news-300")

print("Word2Vec model loaded successfully.")
```

# 🔢 3. Extracting a Word Vector

Once the pretrained model is loaded, the vector representation of a word can be accessed using:

`wv["word"]`

For example:

`wv["king"]`

returns the numerical embedding learned for the word `king`.

Since this model uses **300-dimensional embeddings**, the output contains 300 floating-point values.

These values individually do not have simple human-readable meanings.

Instead, the complete vector represents the word's position in the learned semantic embedding space.

```python
# Retrieve the vector representation of "king"
king_vector = wv["king"]

print("Word: king")
print("Vector Shape:", king_vector.shape)

print("\nFirst 10 Vector Values:")
print(king_vector[:10])
```

# 📐 Understanding the Vector Shape

The shape of the vector can be checked using:

`king_vector.shape`

### Output

`(300,)`

This means that the word `king` is represented using a **one-dimensional array containing 300 numerical features**.

Conceptually:

`king → [x₁, x₂, x₃, ..., x₃₀₀]`

Every word available in this pretrained model has the same embedding dimension.

For example:

`king → 300 dimensions`

`queen → 300 dimensions`

`cricket → 300 dimensions`

Because all words exist in the same vector space, mathematical comparisons can be performed between them.

# 🏏 4. Exploring Another Word Vector

We can retrieve the embedding of any word available in the model's vocabulary.

For example:

`cricket`

The returned vector contains the learned numerical representation of the word based on the context in which it appeared in the model's training data.

```python
# Retrieve the vector representation of "cricket"
cricket_vector = wv["cricket"]

print("Word: cricket")
print("Vector Shape:", cricket_vector.shape)

print("\nFirst 10 Vector Values:")
print(cricket_vector[:10])
```

# 🔍 5. Finding Semantically Similar Words

One of the most useful properties of Word2Vec is its ability to identify words with similar vector representations.

Gensim provides:

`most_similar()`

The method compares vectors in the embedding space and returns words whose vectors are closest to the given word.

The result contains:

`(similar_word, similarity_score)`

The similarity score is based on **cosine similarity**.

A higher score indicates greater similarity in the learned embedding space.

> Similarity in Word2Vec represents similarity in learned usage and context. It does not always mean that two words are exact synonyms.

```python
# Find words most similar to "cricket"
similar_to_cricket = wv.most_similar(
    "cricket",
    topn=10
)

print("Words Most Similar to 'cricket'")
print("-" * 50)

for word, score in similar_to_cricket:
    print(f"{word:<25} {score:.4f}")
```

```python
wv.most_similar("cricket")
```

```python
# Find words most similar to "happy"
similar_to_happy = wv.most_similar(
    "happy",
    topn=10
)

print("Words Most Similar to 'happy'")
print("-" * 50)

for word, score in similar_to_happy:
    print(f"{word:<25} {score:.4f}")
```

# 📏 6. Measuring Similarity Between Two Words

Word2Vec can measure the similarity between two words using:

`wv.similarity(word1, word2)`

The calculation is based on **cosine similarity** between their vectors.

Conceptually:

`Word 1 Vector ↔ Word 2 Vector`

The cosine similarity measures how closely the two vectors point in the same direction.

In general:

- Values closer to `1` indicate stronger similarity.
- Values around `0` indicate weak similarity.
- Negative values can indicate opposing vector directions.

However, these values should be interpreted as relationships learned from the training corpus rather than absolute measures of human semantic similarity.

```python
# Measure semantic similarity
word1 = "hockey"
word2 = "sports"

similarity_score = wv.similarity(
    word1,
    word2
)

print(f"Word 1     : {word1}")
print(f"Word 2     : {word2}")
print(f"Similarity : {similarity_score:.4f}")
```

```python
# Compare different word pairs
word_pairs = [
    ("king", "queen"),
    ("hockey", "sports"),
    ("happy", "joyful"),
    ("car", "vehicle"),
    ("king", "banana"),
]

print("Word Similarity Comparison")
print("=" * 55)

for word1, word2 in word_pairs:

    score = wv.similarity(
        word1,
        word2
    )

    print(
        f"{word1:<12} ↔ "
        f"{word2:<12} : "
        f"{score:.4f}"
    )
```

# ➕ 7. Word Vector Arithmetic

One of the most interesting properties of Word2Vec is that mathematical operations can sometimes capture semantic relationships between words.

A famous example is:

`king - man + woman ≈ queen`

Conceptually:

1. Start with the vector for `king`.
2. Subtract characteristics associated with `man`.
3. Add characteristics associated with `woman`.
4. Search for the word vector closest to the resulting vector.

Mathematically:

`Vector("king") - Vector("man") + Vector("woman")`

The resulting vector may be positioned close to:

`Vector("queen")`

This demonstrates that some relationships between words can emerge as directions in the embedding space.

> Word-vector arithmetic is an approximate property of learned embeddings, not a guaranteed symbolic rule.

```python
# Perform vector arithmetic
result_vector = (
    wv["king"]
    - wv["man"]
    + wv["woman"]
)
print("Result Vector Shape:")
print(result_vector.shape)
```

```python
# Find words closest to the resulting vector
similar_words = wv.most_similar(
    positive=[result_vector],
    topn=10
)

print("Closest Words to: king - man + woman")
print("-" * 55)

for word, score in similar_words:
    print(f"{word:<25} {score:.4f}")
```

# 🎯 8. Performing Word Analogies Directly

Gensim's `most_similar()` can perform vector arithmetic directly using the `positive` and `negative` parameters.

For the relationship:

`king - man + woman`

we can write:

`positive = ["king", "woman"]`

`negative = ["man"]`

Conceptually:

`king + woman - man`

This is usually cleaner than manually constructing the resulting vector.

```python
# Perform the analogy directly
analogy_results = wv.most_similar(
    positive=["king", "woman"],
    negative=["man"],
    topn=10
)

print("Analogy: king - man + woman")
print("=" * 50)

for word, score in analogy_results:
    print(f"{word:<25} {score:.4f}")
```

```python
# Country-capital relationship
capital_analogy = wv.most_similar(
    positive=["France", "Berlin"],
    negative=["Paris"],
    topn=5
)

print("Analogy: France - Paris + Berlin")
print("-" * 50)

for word, score in capital_analogy:
    print(f"{word:<25} {score:.4f}")
```

# ⚠️ 9. Handling Out-of-Vocabulary Words

A pretrained Word2Vec model can only return vectors for words that exist in its vocabulary.

If an unknown word is requested directly, a `KeyError` may occur.

Therefore, before retrieving a word vector, we can check whether the word exists in the model.

```python
def get_word_vector(word):
    """
    Retrieve the vector of a word if it exists
    in the pretrained Word2Vec vocabulary.
    """

    if word in wv.key_to_index:
        return wv[word]

    print(
        f"'{word}' is not available "
        "in the Word2Vec vocabulary."
    )

    return None
```

```python
vector = get_word_vector("king")

if vector is not None:
    print("Vector Shape:", vector.shape)
```

# ⚖️ 10. BoW vs TF-IDF vs Word2Vec

The three techniques studied so far represent text differently.

| Feature | Bag of Words | TF-IDF | Word2Vec |
|---|---|---|---|
| Representation | Sparse | Sparse | Dense |
| Values | Counts / Binary | Importance weights | Learned numerical values |
| Captures semantics | No | Very limited | Yes |
| Fixed word-vector size | No | No | Yes |
| Vocabulary dependent | Yes | Yes | Yes |
| Word similarity | No | Limited | Yes |
| Vector arithmetic | No | No | Yes |
| Context during training | No | No | Yes |

## Bag of Words

Represents:

> How many times does a word appear?

## TF-IDF

Represents:

> How important is a word in this document relative to the corpus?

## Word2Vec

Attempts to represent:

> What semantic relationships does this word have with other words based on usage patterns?

### Representation Example

**BoW**

`king → [0, 1, 0, 0, 1, ...]`

**TF-IDF**

`king → [0.00, 0.72, 0.00, 0.31, ...]`

**Word2Vec**

`king → [0.12, -0.34, 0.78, ..., 0.41]`

The key difference is that Word2Vec produces a **dense learned embedding**, whereas BoW and TF-IDF create features based directly on the corpus vocabulary.

# 🧠 Key Takeaways

- Word2Vec represents words as **dense numerical vectors**.
- Words appearing in similar contexts tend to receive similar vector representations.
- The pretrained model used in this notebook represents each word using **300 dimensions**.
- `wv["word"]` retrieves the vector representation of a word.
- `most_similar()` finds nearby words in the embedding space.
- `similarity()` measures cosine similarity between two word vectors.
- Word2Vec supports meaningful vector arithmetic in some cases.
- Relationships such as `king - man + woman ≈ queen` can emerge from the learned vector space.
- Pretrained models can be used without training Word2Vec from scratch.
- Words missing from the model vocabulary cannot be directly represented.
- Word2Vec produces dense semantic representations, unlike the sparse representations generated by BoW and TF-IDF.
- Word2Vec captures one vector per vocabulary word, regardless of the sentence in which that word appears.
