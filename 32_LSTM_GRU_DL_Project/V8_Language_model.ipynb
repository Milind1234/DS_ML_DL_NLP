```python
# ==============================================================================
# STEP 23.6 — V8 RAW CORPUS FORENSIC AUDIT
# ==============================================================================

print("=" * 80)
print("STEP 23.6 — V8 RAW CORPUS FORENSIC AUDIT")
print("=" * 80)

# ------------------------------------------------------------------------------
# Basic character statistics
# ------------------------------------------------------------------------------

total_chars = len(raw_text)

whitespace_chars = sum(c.isspace() for c in raw_text)
alphabetic_chars = sum(c.isalpha() for c in raw_text)
digit_chars = sum(c.isdigit() for c in raw_text)
punctuation_chars = sum(c in string.punctuation for c in raw_text)

print("\n[1] CHARACTER COMPOSITION")
print("-" * 80)

print(f"Total characters       : {total_chars:,}")
print(f"Whitespace characters  : {whitespace_chars:,}")
print(f"Alphabetic characters  : {alphabetic_chars:,}")
print(f"Digit characters       : {digit_chars:,}")
print(f"Punctuation characters : {punctuation_chars:,}")

# ------------------------------------------------------------------------------
# Line statistics
# ------------------------------------------------------------------------------

lines = raw_text.splitlines()

non_empty_lines = [line for line in lines if line.strip()]

print("\n[2] LINE STATISTICS")
print("-" * 80)

print(f"Total lines            : {len(lines):,}")
print(f"Non-empty lines        : {len(non_empty_lines):,}")
print(f"Empty lines            : {len(lines) - len(non_empty_lines):,}")

# ------------------------------------------------------------------------------
# Whitespace statistics
# ------------------------------------------------------------------------------

print("\n[3] WHITESPACE")
print("-" * 80)

multiple_spaces = len(re.findall(r" {2,}", raw_text))
tabs = raw_text.count("\t")
newlines = raw_text.count("\n")
carriage_returns = raw_text.count("\r")

print(f"Multiple-space regions : {multiple_spaces:,}")
print(f"Tabs                   : {tabs:,}")
print(f"Newline characters     : {newlines:,}")
print(f"Carriage returns       : {carriage_returns:,}")

# ------------------------------------------------------------------------------
# Character inventory
# ------------------------------------------------------------------------------

unique_chars = sorted(set(raw_text))

print("\n[4] CHARACTER INVENTORY")
print("-" * 80)

print(f"Unique characters      : {len(unique_chars)}")

print("\nCharacters:")
print(" ".join(repr(c) for c in unique_chars))

# ------------------------------------------------------------------------------
# Digits
# ------------------------------------------------------------------------------

digits_found = sorted(set(c for c in raw_text if c.isdigit()))

print("\n[5] DIGIT INVENTORY")
print("-" * 80)

print(f"Digits found           : {digits_found}")

# ------------------------------------------------------------------------------
# Punctuation
# ------------------------------------------------------------------------------

punctuation_found = sorted(set(c for c in raw_text if c in string.punctuation))

print("\n[6] PUNCTUATION INVENTORY")
print("-" * 80)

print(f"Unique punctuation characters : {len(punctuation_found)}")
print(" ".join(punctuation_found))

# ------------------------------------------------------------------------------
# Assertions
# ------------------------------------------------------------------------------

assert isinstance(raw_text, str)
assert len(raw_text) > 0
assert total_chars > 0

print("\n[7] VALIDATION")
print("-" * 80)

print("✓ raw_text is a string")
print("✓ raw_text is not empty")
print("✓ Character statistics calculated")
print("✓ Line statistics calculated")
print("✓ Character inventory calculated")

print("\n" + "=" * 80)
print("STEP 23.6 COMPLETED")
print("=" * 80)
```

```python
# ==============================================================================
# STEP 23.7 — HAMLET STRUCTURE INSPECTION
# ==============================================================================

print("=" * 80)
print("STEP 23.7 — HAMLET STRUCTURE INSPECTION")
print("=" * 80)

# ------------------------------------------------------------------------------
# Search for common play metadata / structural markers
# ------------------------------------------------------------------------------

structural_terms = [
    "ACT",
    "SCENE",
    "Enter",
    "Exit",
    "Exeunt",
    "FINIS",
    "HAMLET",
    "KING",
    "QUEEN",
    "POLONIUS",
    "HORATIO",
    "LAERTES",
    "OPHELIA",
]

print("\n[1] STRUCTURAL TERM OCCURRENCES")
print("-" * 80)

lower_text = raw_text.lower()

for term in structural_terms:
    count = len(re.findall(rf"\b{re.escape(term.lower())}\b", lower_text))
    print(f"{term:<15} : {count:,}")

# ------------------------------------------------------------------------------
# Show lines containing structural markers
# ------------------------------------------------------------------------------

print("\n[2] SAMPLE STRUCTURAL LINES")
print("-" * 80)

sample_count = 0

for line_number, line in enumerate(lines, start=1):

    stripped = line.strip()

    if not stripped:
        continue

    line_lower = stripped.lower()

    if any(
        re.search(rf"\b{re.escape(term.lower())}\b", line_lower)
        for term in ["enter", "exit", "exeunt", "actus", "scoena", "scene"]
    ):
        print(f"{line_number:>5}: {stripped[:200]}")

        sample_count += 1

        if sample_count >= 30:
            break

print("\n" + "=" * 80)
print("STEP 23.7 COMPLETED")
print("=" * 80)
```

```python
# ==============================================================================
# STEP 24.1 — V8 NORMALIZATION PREVIEW
# ==============================================================================
# PURPOSE:
#   Compare the raw Hamlet text with a MINIMAL normalization strategy.
#
# IMPORTANT:
#   - This does NOT modify raw_text_v8.
#   - This does NOT create a tokenizer.
#   - This does NOT create vocabulary.
#   - This does NOT create sequences.
#   - This does NOT train anything.
# ==============================================================================

print("=" * 80)
print("STEP 24.1 — V8 NORMALIZATION PREVIEW")
print("=" * 80)

# ------------------------------------------------------------------------------
# Preserve original text
# ------------------------------------------------------------------------------

assert raw_text_v8 == raw_text

# ------------------------------------------------------------------------------
# Minimal normalization
# ------------------------------------------------------------------------------
# We intentionally:
#
#   1. Normalize line endings
#   2. Replace repeated whitespace with a single space
#   3. Strip leading/trailing whitespace
#
# We DO NOT:
#
#   - modernize spelling
#   - remove apostrophes
#   - remove punctuation
#   - remove stage directions
#   - remove character names
#   - remove digits
#   - lowercase yet
# ------------------------------------------------------------------------------

minimal_text_v8 = raw_text_v8.replace("\r\n", "\n").replace("\r", "\n")

minimal_text_v8 = re.sub(r"\s+", " ", minimal_text_v8).strip()

print("\n[1] TEXT SIZE")
print("-" * 80)

print(f"Raw characters       : {len(raw_text_v8):,}")
print(f"Minimal-normalized   : {len(minimal_text_v8):,}")
print(f"Characters removed   : {len(raw_text_v8) - len(minimal_text_v8):,}")

print("\n[2] FIRST 1,000 CHARACTERS AFTER MINIMAL NORMALIZATION")
print("-" * 80)

print(minimal_text_v8[:1000])

# ------------------------------------------------------------------------------
# Safety check
# ------------------------------------------------------------------------------

print("\n[3] SAFETY")
print("-" * 80)

print(f"Original raw_text_v8 unchanged : {raw_text_v8 == raw_text}")
print(f"Minimal text created           : {isinstance(minimal_text_v8, str)}")
print(f"Minimal text non-empty         : {len(minimal_text_v8) > 0}")

assert raw_text_v8 == raw_text
assert isinstance(minimal_text_v8, str)
assert len(minimal_text_v8) > 0

print("\n" + "=" * 80)
print("STEP 24.1 COMPLETED")
print("=" * 80)
```

```python
# ==============================================================================
# STEP 24.2 — V8 CASE NORMALIZATION COMPARISON
# ==============================================================================

print("=" * 80)
print("STEP 24.2 — V8 CASE NORMALIZATION COMPARISON")
print("=" * 80)

# ------------------------------------------------------------------------------
# Candidate A: Preserve case
# ------------------------------------------------------------------------------

case_preserved_text_v8 = minimal_text_v8

# ------------------------------------------------------------------------------
# Candidate B: Lowercase
# ------------------------------------------------------------------------------

lowercase_text_v8 = minimal_text_v8.lower()

# ------------------------------------------------------------------------------
# Character statistics
# ------------------------------------------------------------------------------

upper_chars = sum(c.isupper() for c in minimal_text_v8)
lower_chars = sum(c.islower() for c in minimal_text_v8)

print("\n[1] CASE STATISTICS")
print("-" * 80)

print(f"Uppercase alphabetic characters : {upper_chars:,}")
print(f"Lowercase alphabetic characters : {lower_chars:,}")

print("\n[2] TEXT LENGTH")
print("-" * 80)

print(f"Case-preserved characters : {len(case_preserved_text_v8):,}")
print(f"Lowercase characters      : {len(lowercase_text_v8):,}")

# ------------------------------------------------------------------------------
# Important examples
# ------------------------------------------------------------------------------

case_examples = [
    "The",
    "the",
    "Actus",
    "actus",
    "Enter",
    "enter",
    "Hamlet",
    "HAMLET",
    "Who's",
    "who's",
    "Haue",
    "haue",
]

print("\n[3] CASE EXAMPLES")
print("-" * 80)

for example in case_examples:
    preserved_count = minimal_text_v8.count(example)
    lowercase_count = lowercase_text_v8.count(example.lower())

    print(
        f"{example:<10} | "
        f"preserved exact count = {preserved_count:<5} | "
        f"lowercase form count = {lowercase_count:<5}"
    )

# ------------------------------------------------------------------------------
# Decision preview
# ------------------------------------------------------------------------------

print("\n[4] INITIAL DESIGN CONSIDERATION")
print("-" * 80)

print("Case preservation increases the number of distinct surface forms.")
print("Lowercasing allows 'The' and 'the' to share representation.")
print()
print("For a small corpus, reducing unnecessary vocabulary fragmentation")
print("is generally desirable.")
print()
print("FINAL CASE DECISION WILL BE FROZEN AFTER TOKENIZATION AUDIT.")

print("\n" + "=" * 80)
print("STEP 24.2 COMPLETED")
print("=" * 80)
```

```python
# ==============================================================================
# STEP 24.3 — V8 HISTORICAL SPELLING PRESERVATION TEST
# ==============================================================================

print("=" * 80)
print("STEP 24.3 — V8 HISTORICAL SPELLING PRESERVATION TEST")
print("=" * 80)

historical_examples = [
    "haue",
    "vpon",
    "vnfold",
    "selfe",
    "giue",
    "giuing",
    "vndertake",
    "heere",
    "beleeue",
    "relieu'd",
    "sicke",
    "strook",
    "thankes",
    "releefe",
]

print("\n[1] HISTORICAL / RARE FORMS")
print("-" * 80)

for word in historical_examples:

    exact_count = len(
        re.findall(
            rf"(?<!\w){re.escape(word)}(?!\w)",
            minimal_text_v8,
            flags=re.IGNORECASE
        )
    )

    print(f"{word:<15} : {exact_count:>4}")

# ------------------------------------------------------------------------------
# Explicit modernization check
# ------------------------------------------------------------------------------

modernization_map = {
    "haue": "have",
    "vpon": "upon",
    "vnfold": "unfold",
    "selfe": "self",
    "giue": "give",
    "vndertake": "undertake",
    "heere": "here",
    "beleeue": "believe",
}

print("\n[2] IMPORTANT DESIGN RULE")
print("-" * 80)

print("We will NOT manually replace historical spellings.")

for old_word, modern_word in modernization_map.items():
    print(f"Preserve: {old_word:<12} | Do NOT force: {modern_word}")

print("\nReason:")
print("Historical forms are part of the original corpus.")
print("V8 should test whether subword tokenization can represent them")
print("without manually changing the source text.")

print("\n" + "=" * 80)
print("STEP 24.3 COMPLETED")
print("=" * 80)
```

```python
# ==============================================================================
# STEP 24.4 — V8 APOSTROPHE / CONTRACTION AUDIT
# ==============================================================================

print("=" * 80)
print("STEP 24.4 — V8 APOSTROPHE / CONTRACTION AUDIT")
print("=" * 80)

apostrophe_patterns = [
    r"\b\w+'\w+\b",
    r"\b'\w+\b",
    r"\w+'\b",
]

apostrophe_matches = []

for pattern in apostrophe_patterns:
    apostrophe_matches.extend(
        re.findall(pattern, minimal_text_v8)
    )

# Unique forms
apostrophe_counter = collections.Counter(apostrophe_matches)

print("\n[1] APOSTROPHE-CONTAINING FORMS")
print("-" * 80)

print(f"Total apostrophe-containing matches : {len(apostrophe_matches):,}")
print(f"Unique apostrophe forms             : {len(apostrophe_counter):,}")

print("\nTop 40 forms:")

for word, count in apostrophe_counter.most_common(40):
    print(f"{word:<20} : {count}")

# ------------------------------------------------------------------------------
# Critical examples
# ------------------------------------------------------------------------------

critical_apostrophe_examples = [
    "Who's",
    "who's",
    "'Tis",
    "'tis",
    "ha's",
    "i'th",
    "relieu'd",
    "drown'd",
    "woo't",
    "o're",
    "a'th",
]

print("\n[2] CRITICAL HISTORICAL / CONTRACTION FORMS")
print("-" * 80)

for word in critical_apostrophe_examples:

    count = len(
        re.findall(
            rf"(?<!\w){re.escape(word)}(?!\w)",
            minimal_text_v8,
            flags=re.IGNORECASE
        )
    )

    print(f"{word:<15} : {count:>4}")

print("\n[3] DESIGN DECISION")
print("-" * 80)

print("✓ Apostrophes will be preserved.")
print("✓ Contraction-like forms will remain intact at the source-text level.")
print("✓ We will let the subword tokenizer decide how to split them.")

print("\n" + "=" * 80)
print("STEP 24.4 COMPLETED")
print("=" * 80)
```

```python
# ==============================================================================
# STEP 24.5 — V8 PUNCTUATION PRESERVATION AUDIT
# ==============================================================================

print("=" * 80)
print("STEP 24.5 — V8 PUNCTUATION PRESERVATION AUDIT")
print("=" * 80)

punctuation_counter = collections.Counter(
    char
    for char in minimal_text_v8
    if char in string.punctuation
)

print("\n[1] PUNCTUATION FREQUENCY")
print("-" * 80)

for punctuation, count in punctuation_counter.most_common():
    print(f"{repr(punctuation):<8} : {count:,}")

print("\n[2] DESIGN DECISION")
print("-" * 80)

print("Punctuation will NOT be globally deleted.")
print()
print("Reason:")
print("Punctuation provides syntactic and sentence-boundary information.")
print("A subword tokenizer can learn appropriate representations for it.")

print("\n" + "=" * 80)
print("STEP 24.5 COMPLETED")
print("=" * 80)
```

```python
# ============================================================
# STEP 25.3 — TRAIN BPE TOKENIZER CANDIDATES
# ============================================================

from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import ByteLevel
from tokenizers.decoders import ByteLevel as ByteLevelDecoder

# ------------------------------------------------------------
# V8 tokenizer configuration
# ------------------------------------------------------------

PAD_TOKEN = "<PAD>"
UNK_TOKEN = "<UNK>"

TARGET_VOCAB_SIZES = [2000, 2500, 3000]

MIN_SUBWORD_FREQUENCY = 2

v8_bpe_candidates = {}

print("=" * 70)
print("V8 BPE TOKENIZER TRAINING")
print("=" * 70)

print("Training corpus : TRAIN ONLY")
print(f"Minimum frequency : {MIN_SUBWORD_FREQUENCY}")
print(f"Candidate vocabularies : {TARGET_VOCAB_SIZES}")

for target_vocab_size in TARGET_VOCAB_SIZES:

    print("\n" + "-" * 70)
    print(f"Training BPE tokenizer → target vocab = {target_vocab_size}")

    # --------------------------------------------------------
    # BPE model
    # --------------------------------------------------------

    tokenizer = Tokenizer(
        BPE(
            unk_token=UNK_TOKEN
        )
    )

    # --------------------------------------------------------
    # Byte-Level pre-tokenizer
    #
    # Advantages:
    # - preserves punctuation
    # - handles apostrophes
    # - handles unseen words
    # - preserves whitespace information
    # - works well with historical spelling
    # --------------------------------------------------------

    tokenizer.pre_tokenizer = ByteLevel(
        add_prefix_space=True
    )

    tokenizer.decoder = ByteLevelDecoder()

    # --------------------------------------------------------
    # Trainer
    # --------------------------------------------------------

    trainer = BpeTrainer(
        vocab_size=target_vocab_size,
        min_frequency=MIN_SUBWORD_FREQUENCY,
        special_tokens=[
            PAD_TOKEN,
            UNK_TOKEN
        ],
        show_progress=True
    )

    # --------------------------------------------------------
    # IMPORTANT:
    # Train ONLY on training text.
    # --------------------------------------------------------

    tokenizer.train_from_iterator(
        [train_text_v8],
        trainer=trainer
    )

    actual_vocab_size = tokenizer.get_vocab_size(
        with_added_tokens=True
    )

    pad_id = tokenizer.token_to_id(PAD_TOKEN)
    unk_id = tokenizer.token_to_id(UNK_TOKEN)

    v8_bpe_candidates[target_vocab_size] = tokenizer

    print(f"Target vocabulary : {target_vocab_size:,}")
    print(f"Actual vocabulary : {actual_vocab_size:,}")
    print(f"PAD ID            : {pad_id}")
    print(f"UNK ID            : {unk_id}")

    print("✅ Tokenizer trained successfully.")

print("\n" + "=" * 70)
print("ALL BPE CANDIDATES TRAINED")
print("=" * 70)

print(
    "Candidates:",
    list(v8_bpe_candidates.keys())
)
```

```python
# ============================================================
# STEP 25.4 — SUBWORD TOKENIZATION STATISTICS
# ============================================================

import numpy as np
import pandas as pd

def analyze_split_tokenization(tokenizer, text):
    """
    Analyze the complete split using the trained tokenizer.
    """

    encoding = tokenizer.encode(
        text,
        add_special_tokens=False
    )

    ids = encoding.ids
    tokens = encoding.tokens

    unk_id = tokenizer.token_to_id(UNK_TOKEN)

    unk_count = sum(
        token_id == unk_id
        for token_id in ids
    )

    total_tokens = len(ids)

    unk_rate = (
        unk_count / total_tokens * 100
        if total_tokens > 0
        else 0
    )

    return {
        "subword_tokens": total_tokens,
        "unk_tokens": unk_count,
        "unk_rate_percent": unk_rate,
        "unique_subword_ids": len(set(ids))
    }


results = []

for target_vocab_size, tokenizer in v8_bpe_candidates.items():

    train_stats = analyze_split_tokenization(
        tokenizer,
        train_text_v8
    )

    val_stats = analyze_split_tokenization(
        tokenizer,
        val_text_v8
    )

    test_stats = analyze_split_tokenization(
        tokenizer,
        test_text_v8
    )

    results.append({
        "Target Vocab": target_vocab_size,

        "Actual Vocab":
            tokenizer.get_vocab_size(
                with_added_tokens=True
            ),

        "Train Subwords":
            train_stats["subword_tokens"],

        "Val Subwords":
            val_stats["subword_tokens"],

        "Test Subwords":
            test_stats["subword_tokens"],

        "Train UNK %":
            train_stats["unk_rate_percent"],

        "Val UNK %":
            val_stats["unk_rate_percent"],

        "Test UNK %":
            test_stats["unk_rate_percent"],

        "Train Unique IDs":
            train_stats["unique_subword_ids"],

        "Val Unique IDs":
            val_stats["unique_subword_ids"],

        "Test Unique IDs":
            test_stats["unique_subword_ids"]
    })


v8_tokenization_results = pd.DataFrame(results)

print("=" * 100)
print("V8 SUBWORD TOKENIZATION RESULTS")
print("=" * 100)

display(
    v8_tokenization_results
)
```

```python
# ============================================================
# STEP 25.5 — WORD → SUBWORD INFLATION ANALYSIS
# ============================================================

def analyze_word_inflation(tokenizer, words, sample_limit=None):

    if sample_limit is not None:
        words_to_analyze = words[:sample_limit]
    else:
        words_to_analyze = words

    encodings = tokenizer.encode_batch(
        words_to_analyze,
        add_special_tokens=False
    )

    lengths = np.array([
        len(enc.ids)
        for enc in encodings
    ])

    return {
        "words": len(words_to_analyze),
        "total_subwords": int(lengths.sum()),
        "mean_subwords_per_word": float(lengths.mean()),
        "median_subwords_per_word": float(np.median(lengths)),
        "max_subwords_per_word": int(lengths.max()),
        "p90": float(np.percentile(lengths, 90)),
        "p95": float(np.percentile(lengths, 95)),
        "p99": float(np.percentile(lengths, 99))
    }


inflation_results = []

for target_vocab_size, tokenizer in v8_bpe_candidates.items():

    stats = analyze_word_inflation(
        tokenizer,
        train_words_v8
    )

    inflation_results.append({
        "Target Vocab": target_vocab_size,
        "Train Words": stats["words"],
        "Total Subwords": stats["total_subwords"],
        "Mean Pieces / Word":
            stats["mean_subwords_per_word"],
        "Median Pieces / Word":
            stats["median_subwords_per_word"],
        "Max Pieces / Word":
            stats["max_subwords_per_word"],
        "P90":
            stats["p90"],
        "P95":
            stats["p95"],
        "P99":
            stats["p99"]
    })


v8_inflation_results = pd.DataFrame(
    inflation_results
)

print("=" * 100)
print("V8 WORD → SUBWORD INFLATION")
print("=" * 100)

display(
    v8_inflation_results
)
```

```python
# ============================================================
# STEP 26.3 — TOKEN ID INTEGRITY AUDIT
# ============================================================

def audit_token_ids_v8(name, token_ids):

    print("\n" + "-" * 70)
    print(f"{name} TOKEN ID AUDIT")
    print("-" * 70)

    # --------------------------------------------------------
    # Basic properties
    # --------------------------------------------------------

    assert len(token_ids) > 0
    assert token_ids.dtype == np.int32

    # --------------------------------------------------------
    # Valid ID range
    # --------------------------------------------------------

    min_id = int(token_ids.min())
    max_id = int(token_ids.max())

    valid_range = (
        min_id >= 0
        and max_id < VOCAB_SIZE_V8
    )

    print(f"Minimum ID       : {min_id}")
    print(f"Maximum ID       : {max_id}")
    print(f"Valid ID range   : {valid_range}")

    assert valid_range

    # --------------------------------------------------------
    # NaN / invalid values
    # --------------------------------------------------------

    print(
        f"NaN present      : {np.isnan(token_ids).any()}"
    )

    assert not np.isnan(token_ids).any()

    # --------------------------------------------------------
    # UNK count
    # --------------------------------------------------------

    unk_count = int(
        np.sum(token_ids == UNK_ID_V8)
    )

    unk_rate = (
        unk_count / len(token_ids) * 100
    )

    print(f"UNK tokens       : {unk_count:,}")
    print(f"UNK rate         : {unk_rate:.4f}%")

    return {
        "tokens": len(token_ids),
        "min_id": min_id,
        "max_id": max_id,
        "unk_count": unk_count,
        "unk_rate": unk_rate
    }


train_id_audit = audit_token_ids_v8(
    "TRAIN",
    train_ids_v8
)

val_id_audit = audit_token_ids_v8(
    "VALIDATION",
    val_ids_v8
)

test_id_audit = audit_token_ids_v8(
    "TEST",
    test_ids_v8
)

print("\n" + "=" * 70)
print("✅ TOKEN ID INTEGRITY PASSED")
print("=" * 70)
```

```python
# ============================================================
# STEP 26.4 — ENCODE → DECODE RECONSTRUCTION TEST
# ============================================================

def reconstruction_test_v8(name, original_text):

    encoding = tokenizer_v8.encode(
        original_text,
        add_special_tokens=False
    )

    decoded_text = tokenizer_v8.decode(
        encoding.ids,
        skip_special_tokens=False
    )

    # ByteLevel with add_prefix_space=True
    # can introduce one leading space.
    original_normalized = original_text.strip()
    decoded_normalized = decoded_text.strip()

    exact_match = (
        original_normalized == decoded_normalized
    )

    print("\n" + "-" * 70)
    print(f"{name} RECONSTRUCTION")
    print("-" * 70)

    print(f"Original chars : {len(original_text):,}")
    print(f"Decoded chars  : {len(decoded_text):,}")
    print(f"Exact match    : {exact_match}")

    if not exact_match:

        # Find first mismatch
        mismatch_index = None

        max_len = min(
            len(original_normalized),
            len(decoded_normalized)
        )

        for i in range(max_len):
            if (
                original_normalized[i]
                != decoded_normalized[i]
            ):
                mismatch_index = i
                break

        if mismatch_index is not None:
            start = max(0, mismatch_index - 40)
            end = mismatch_index + 40

            print("\nFirst mismatch around:")
            print(
                "Original:",
                repr(original_normalized[start:end])
            )
            print(
                "Decoded :",
                repr(decoded_normalized[start:end])
            )

    return exact_match


train_reconstruction = reconstruction_test_v8(
    "TRAIN",
    train_text_v8
)

val_reconstruction = reconstruction_test_v8(
    "VALIDATION",
    val_text_v8
)

test_reconstruction = reconstruction_test_v8(
    "TEST",
    test_text_v8
)

print("\n" + "=" * 70)

if (
    train_reconstruction
    and val_reconstruction
    and test_reconstruction
):
    print("✅ ALL RECONSTRUCTION TESTS PASSED")
else:
    print("⚠️ RECONSTRUCTION ISSUE DETECTED")

print("=" * 70)
```

```python
# ============================================================
# STEP 26.7 — CREATE V8 NEXT-TOKEN SEQUENCES
# ============================================================

SEQ_LEN_V8 = 30

def create_sequences_v8(token_ids, seq_len):
    """
    Create next-token prediction sequences.

    X[i] = token_ids[i : i + seq_len]
    y[i] = token_ids[i + seq_len]
    """

    n_sequences = len(token_ids) - seq_len

    if n_sequences <= 0:
        raise ValueError(
            "Not enough tokens to create sequences."
        )

    X = np.zeros(
        (n_sequences, seq_len),
        dtype=np.int32
    )

    y = np.zeros(
        n_sequences,
        dtype=np.int32
    )

    for i in range(n_sequences):

        X[i] = token_ids[
            i : i + seq_len
        ]

        y[i] = token_ids[
            i + seq_len
        ]

    return X, y


X_train_v8, y_train_v8 = create_sequences_v8(
    train_ids_v8,
    SEQ_LEN_V8
)

X_val_v8, y_val_v8 = create_sequences_v8(
    val_ids_v8,
    SEQ_LEN_V8
)

X_test_v8, y_test_v8 = create_sequences_v8(
    test_ids_v8,
    SEQ_LEN_V8
)

print("=" * 80)
print("V8 SEQUENCE DATASETS")
print("=" * 80)

print(
    f"Train X : {X_train_v8.shape}"
)

print(
    f"Train y : {y_train_v8.shape}"
)

print(
    f"Val X   : {X_val_v8.shape}"
)

print(
    f"Val y   : {y_val_v8.shape}"
)

print(
    f"Test X  : {X_test_v8.shape}"
)

print(
    f"Test y  : {y_test_v8.shape}"
)
```

```python
# ============================================================
# STEP 26.10 — HUMAN-READABLE SEQUENCE INSPECTION
# ============================================================

def inspect_sequence_v8(
    tokenizer,
    X,
    y,
    index=0
):

    input_ids = X[index].tolist()
    target_id = int(y[index])

    input_tokens = [
        tokenizer.id_to_token(
            int(token_id)
        )
        for token_id in input_ids
    ]

    target_token = tokenizer.id_to_token(
        target_id
    )

    decoded_context = tokenizer.decode(
        input_ids,
        skip_special_tokens=False
    )

    decoded_target = tokenizer.decode(
        [target_id],
        skip_special_tokens=False
    )

    print("=" * 90)
    print(f"SEQUENCE INDEX = {index}")
    print("=" * 90)

    print("\nInput IDs:")
    print(input_ids)

    print("\nInput subwords:")
    print(input_tokens)

    print("\nDecoded context:")
    print(repr(decoded_context))

    print("\nTarget ID:")
    print(target_id)

    print("\nTarget subword:")
    print(repr(target_token))

    print("\nDecoded target:")
    print(repr(decoded_target))

    print("\nFull context + target:")
    print(
        repr(
            decoded_context + decoded_target
        )
    )


inspect_sequence_v8(
    tokenizer_v8,
    X_train_v8,
    y_train_v8,
    index=0
)
```

```python
# ============================================================
# STEP 27.1 — V8 MODEL CONFIGURATION
# ============================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# ------------------------------------------------------------
# Reproducibility
# ------------------------------------------------------------

SEED = 42

tf.random.set_seed(SEED)
np.random.seed(SEED)

# ------------------------------------------------------------
# V8 Model Configuration
# ------------------------------------------------------------

VOCAB_SIZE_V8 = 2500
SEQ_LEN_V8 = 30

EMBEDDING_DIM_V8 = 128
LSTM_UNITS_V8 = 128
DROPOUT_RATE_V8 = 0.30

BATCH_SIZE_V8 = 64
LEARNING_RATE_V8 = 0.001

MODEL_NAME_V8 = "lstm_v8"

print("=" * 80)
print("V8 MODEL CONFIGURATION")
print("=" * 80)

config_v8 = {
    "Model": MODEL_NAME_V8,
    "Vocabulary Size": VOCAB_SIZE_V8,
    "Sequence Length": SEQ_LEN_V8,
    "Embedding Dimension": EMBEDDING_DIM_V8,
    "LSTM Units": LSTM_UNITS_V8,
    "Dropout": DROPOUT_RATE_V8,
    "Batch Size": BATCH_SIZE_V8,
    "Learning Rate": LEARNING_RATE_V8,
    "Seed": SEED
}

for key, value in config_v8.items():
    print(f"{key:<25}: {value}")

print("\n✅ Model configuration frozen.")
```

```python
# ============================================================
# STEP 27.2 — BUILD V8 LSTM MODEL
# ============================================================

lstm_v8 = keras.Sequential(
    [
        layers.Input(
            shape=(SEQ_LEN_V8,),
            dtype=tf.int32,
            name="input_tokens"
        ),

        layers.Embedding(
            input_dim=VOCAB_SIZE_V8,
            output_dim=EMBEDDING_DIM_V8,
            name="embedding"
        ),

        layers.LSTM(
            LSTM_UNITS_V8,
            name="lstm"
        ),

        layers.Dropout(
            DROPOUT_RATE_V8,
            name="dropout"
        ),

        layers.Dense(
            VOCAB_SIZE_V8,
            activation="softmax",
            name="next_token"
        )
    ],
    name=MODEL_NAME_V8
)

print("=" * 80)
print("V8 LSTM MODEL CREATED")
print("=" * 80)

lstm_v8.summary()
```

```python
# =============================================================================
# STEP 31 — V8 FORENSIC EVALUATION
# =============================================================================
# Purpose:
#   Evaluate the BEST restored V8 model (Epoch 9) without retraining.
#
# Checks:
#   1. Train / Validation / Test loss
#   2. Top-1 / Top-3 / Top-5 / Top-10 accuracy
#   3. Perplexity
#   4. Prediction diversity
#   5. Most frequent predicted tokens
#   6. Actual vs predicted token distribution
#   7. Prediction entropy
#   8. V7-style prediction collapse
#
# IMPORTANT:
#   - No training
#   - No model modification
#   - Test set is used ONLY for evaluation
# =============================================================================

print("=" * 100)
print("🔬 STEP 31 — V8 FORENSIC EVALUATION")
print("=" * 100)

import numpy as np
import pandas as pd
import math
from collections import Counter

# =============================================================================
# 31.1 — RUNTIME SAFETY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("31.1 — RUNTIME SAFETY CHECK")
print("=" * 100)

required_variables = [
    "lstm_v8",
    "X_train_v8",
    "y_train_v8",
    "X_val_v8",
    "y_val_v8",
    "X_test_v8",
    "y_test_v8",
    "tokenizer_v8",
    "VOCAB_SIZE_V8",
    "SEQ_LEN_V8",
    "PAD_ID_V8",
    "UNK_ID_V8",
]

missing_variables = [
    var for var in required_variables
    if var not in globals()
]

if missing_variables:
    raise RuntimeError(
        f"❌ Missing required variables: {missing_variables}"
    )

print("✅ All required V8 variables are available.")

print(f"Model              : {lstm_v8.name}")
print(f"Vocabulary         : {VOCAB_SIZE_V8:,}")
print(f"Sequence length    : {SEQ_LEN_V8}")
print(f"Train samples      : {len(X_train_v8):,}")
print(f"Validation samples : {len(X_val_v8):,}")
print(f"Test samples       : {len(X_test_v8):,}")


# =============================================================================
# 31.2 — CONFIRM BEST MODEL IS LOADED
# =============================================================================

print("\n" + "=" * 100)
print("31.2 — BEST MODEL CHECK")
print("=" * 100)

print("""
The training process restored the weights from the best validation-loss epoch.

Best epoch:
    Epoch 9

Best validation loss:
    5.74377

We will evaluate the current `lstm_v8` weights directly.
""")

print("Model object:", lstm_v8.name)
print("Optimizer:", type(lstm_v8.optimizer).__name__)
print("Loss:", type(lstm_v8.loss).__name__)

print("\n✅ Current model is the restored best model from training.")


# =============================================================================
# 31.3 — BASIC MODEL EVALUATION
# =============================================================================

print("\n" + "=" * 100)
print("31.3 — BASIC LOSS / ACCURACY EVALUATION")
print("=" * 100)

print("\nEvaluating TRAIN set...")

train_loss, train_acc = lstm_v8.evaluate(
    X_train_v8,
    y_train_v8,
    batch_size=BATCH_SIZE_V8,
    verbose=0
)

print("Evaluating VALIDATION set...")

val_loss, val_acc = lstm_v8.evaluate(
    X_val_v8,
    y_val_v8,
    batch_size=BATCH_SIZE_V8,
    verbose=0
)

print("Evaluating TEST set...")

test_loss, test_acc = lstm_v8.evaluate(
    X_test_v8,
    y_test_v8,
    batch_size=BATCH_SIZE_V8,
    verbose=0
)

print("\n" + "-" * 100)
print("BASIC EVALUATION RESULTS")
print("-" * 100)

print(f"{'Dataset':<15} {'Loss':>12} {'Accuracy':>15}")
print("-" * 45)
print(f"{'Train':<15} {train_loss:>12.6f} {train_acc * 100:>14.4f}%")
print(f"{'Validation':<15} {val_loss:>12.6f} {val_acc * 100:>14.4f}%")
print(f"{'Test':<15} {test_loss:>12.6f} {test_acc * 100:>14.4f}%")


# =============================================================================
# 31.4 — PERPLEXITY
# =============================================================================

print("\n" + "=" * 100)
print("31.4 — PERPLEXITY")
print("=" * 100)

train_perplexity = math.exp(train_loss)
val_perplexity = math.exp(val_loss)
test_perplexity = math.exp(test_loss)

print(f"Train perplexity      : {train_perplexity:.4f}")
print(f"Validation perplexity : {val_perplexity:.4f}")
print(f"Test perplexity       : {test_perplexity:.4f}")

print("""
Interpretation:
    Lower perplexity = better language-model prediction.

    Random/uniform perplexity for a 2,500-token vocabulary would be
    approximately 2,500.

    Therefore, we expect a trained model to be substantially below 2,500.
""")


# =============================================================================
# 31.5 — GENERATE PREDICTION MATRICES
# =============================================================================

print("\n" + "=" * 100)
print("31.5 — GENERATING PREDICTIONS")
print("=" * 100)

print("Generating TRAIN predictions...")
train_probs = lstm_v8.predict(
    X_train_v8,
    batch_size=BATCH_SIZE_V8,
    verbose=0
)

print("Generating VALIDATION predictions...")
val_probs = lstm_v8.predict(
    X_val_v8,
    batch_size=BATCH_SIZE_V8,
    verbose=0
)

print("Generating TEST predictions...")
test_probs = lstm_v8.predict(
    X_test_v8,
    batch_size=BATCH_SIZE_V8,
    verbose=0
)

print("\nPrediction shapes:")
print(f"Train      : {train_probs.shape}")
print(f"Validation : {val_probs.shape}")
print(f"Test       : {test_probs.shape}")

expected_train_shape = (len(X_train_v8), VOCAB_SIZE_V8)
expected_val_shape = (len(X_val_v8), VOCAB_SIZE_V8)
expected_test_shape = (len(X_test_v8), VOCAB_SIZE_V8)

assert train_probs.shape == expected_train_shape
assert val_probs.shape == expected_val_shape
assert test_probs.shape == expected_test_shape

print("\n✅ Prediction shapes are valid.")


# =============================================================================
# 31.6 — PREDICTED TOKEN IDs
# =============================================================================

print("\n" + "=" * 100)
print("31.6 — PREDICTED TOKEN IDs")
print("=" * 100)

train_pred_ids = np.argmax(train_probs, axis=1)
val_pred_ids = np.argmax(val_probs, axis=1)
test_pred_ids = np.argmax(test_probs, axis=1)

print("✅ Argmax predictions generated.")

print("\nPrediction ID ranges:")
print(f"Train      : {train_pred_ids.min()} → {train_pred_ids.max()}")
print(f"Validation : {val_pred_ids.min()} → {val_pred_ids.max()}")
print(f"Test       : {test_pred_ids.min()} → {test_pred_ids.max()}")


# =============================================================================
# 31.7 — TOKEN ID → TOKEN HELPER
# =============================================================================

print("\n" + "=" * 100)
print("31.7 — TOKEN DECODING HELPER")
print("=" * 100)

# HuggingFace tokenizers supports id_to_token()
def decode_token_id(token_id):
    try:
        token = tokenizer_v8.id_to_token(int(token_id))
        return token if token is not None else "<UNKNOWN_ID>"
    except Exception:
        return "<UNKNOWN_ID>"

print("Sample vocabulary tokens:")

sample_ids = [
    0,
    1,
    7,
    9,
    55,
    71,
    78,
    81,
    101
]

for token_id in sample_ids:
    if token_id < VOCAB_SIZE_V8:
        print(
            f"ID {token_id:4d} → {decode_token_id(token_id)!r}"
        )

print("\n✅ Token decoding helper ready.")


# =============================================================================
# 31.8 — TOP-K ACCURACY
# =============================================================================

print("\n" + "=" * 100)
print("31.8 — TOP-K ACCURACY")
print("=" * 100)

def calculate_top_k_accuracy(probs, targets, k):
    """
    Calculate Top-K categorical accuracy.

    A prediction is correct if the actual target token
    appears anywhere inside the model's top-K predictions.
    """

    top_k_ids = np.argpartition(
        probs,
        -k,
        axis=1
    )[:, -k:]

    correct = np.any(
        top_k_ids == targets[:, None],
        axis=1
    )

    return np.mean(correct)


k_values = [1, 3, 5, 10]

results_topk = {}

for k in k_values:

    train_topk = calculate_top_k_accuracy(
        train_probs,
        y_train_v8,
        k
    )

    val_topk = calculate_top_k_accuracy(
        val_probs,
        y_val_v8,
        k
    )

    test_topk = calculate_top_k_accuracy(
        test_probs,
        y_test_v8,
        k
    )

    results_topk[k] = {
        "Train": train_topk,
        "Validation": val_topk,
        "Test": test_topk
    }

print("\nTop-K accuracy:")
print("-" * 70)
print(
    f"{'K':<8}"
    f"{'Train':>15}"
    f"{'Validation':>18}"
    f"{'Test':>15}"
)
print("-" * 70)

for k in k_values:

    print(
        f"Top-{k:<3}"
        f"{results_topk[k]['Train'] * 100:>14.4f}%"
        f"{results_topk[k]['Validation'] * 100:>17.4f}%"
        f"{results_topk[k]['Test'] * 100:>14.4f}%"
    )


# =============================================================================
# 31.9 — PREDICTION DIVERSITY
# =============================================================================

print("\n" + "=" * 100)
print("31.9 — PREDICTION DIVERSITY")
print("=" * 100)

def prediction_diversity(pred_ids, vocab_size):
    unique_ids = np.unique(pred_ids)

    return {
        "unique_count": len(unique_ids),
        "unique_percentage": len(unique_ids) / vocab_size * 100,
        "total_predictions": len(pred_ids)
    }


train_diversity = prediction_diversity(
    train_pred_ids,
    VOCAB_SIZE_V8
)

val_diversity = prediction_diversity(
    val_pred_ids,
    VOCAB_SIZE_V8
)

test_diversity = prediction_diversity(
    test_pred_ids,
    VOCAB_SIZE_V8
)

print("\nPrediction diversity:")
print("-" * 75)

print(
    f"Train      : "
    f"{train_diversity['unique_count']} unique tokens "
    f"({train_diversity['unique_percentage']:.2f}% of vocabulary)"
)

print(
    f"Validation : "
    f"{val_diversity['unique_count']} unique tokens "
    f"({val_diversity['unique_percentage']:.2f}% of vocabulary)"
)

print(
    f"Test       : "
    f"{test_diversity['unique_count']} unique tokens "
    f"({test_diversity['unique_percentage']:.2f}% of vocabulary)"
)


# =============================================================================
# 31.10 — MOST FREQUENT PREDICTIONS
# =============================================================================

print("\n" + "=" * 100)
print("31.10 — MOST FREQUENT PREDICTED TOKENS")
print("=" * 100)

def show_prediction_distribution(pred_ids, split_name, top_n=20):

    counter = Counter(pred_ids)

    print(f"\n{split_name.upper()} — TOP {top_n} PREDICTIONS")
    print("-" * 80)

    print(
        f"{'Rank':<7}"
        f"{'ID':<8}"
        f"{'Token':<20}"
        f"{'Count':<12}"
        f"{'Percentage':<12}"
    )

    print("-" * 80)

    total = len(pred_ids)

    for rank, (token_id, count) in enumerate(
        counter.most_common(top_n),
        start=1
    ):

        token = decode_token_id(token_id)
        percentage = count / total * 100

        print(
            f"{rank:<7}"
            f"{token_id:<8}"
            f"{str(token)[:18]:<20}"
            f"{count:<12}"
            f"{percentage:<11.4f}%"
        )


show_prediction_distribution(
    train_pred_ids,
    "Train"
)

show_prediction_distribution(
    val_pred_ids,
    "Validation"
)

show_prediction_distribution(
    test_pred_ids,
    "Test"
)


# =============================================================================
# 31.11 — ACTUAL TARGET DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("31.11 — ACTUAL TARGET DISTRIBUTION")
print("=" * 100)

def show_actual_distribution(target_ids, split_name, top_n=20):

    counter = Counter(target_ids)

    print(f"\n{split_name.upper()} — TOP {top_n} ACTUAL TARGETS")
    print("-" * 80)

    print(
        f"{'Rank':<7}"
        f"{'ID':<8}"
        f"{'Token':<20}"
        f"{'Count':<12}"
        f"{'Percentage':<12}"
    )

    print("-" * 80)

    total = len(target_ids)

    for rank, (token_id, count) in enumerate(
        counter.most_common(top_n),
        start=1
    ):

        token = decode_token_id(token_id)
        percentage = count / total * 100

        print(
            f"{rank:<7}"
            f"{token_id:<8}"
            f"{str(token)[:18]:<20}"
            f"{count:<12}"
            f"{percentage:<11.4f}%"
        )


show_actual_distribution(
    y_train_v8,
    "Train"
)

show_actual_distribution(
    y_val_v8,
    "Validation"
)

show_actual_distribution(
    y_test_v8,
    "Test"
)


# =============================================================================
# 31.12 — ACTUAL VS PREDICTED DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("31.12 — ACTUAL VS PREDICTED DISTRIBUTION")
print("=" * 100)

def distribution_comparison(targets, predictions, split_name, top_n=20):

    actual_counter = Counter(targets)
    pred_counter = Counter(predictions)

    all_ids = set(
        list(actual_counter.keys()) +
        list(pred_counter.keys())
    )

    rows = []

    for token_id in all_ids:

        actual_count = actual_counter.get(token_id, 0)
        pred_count = pred_counter.get(token_id, 0)

        rows.append({
            "id": token_id,
            "token": decode_token_id(token_id),
            "actual_count": actual_count,
            "predicted_count": pred_count,
            "actual_pct": actual_count / len(targets) * 100,
            "predicted_pct": pred_count / len(predictions) * 100
        })

    df = pd.DataFrame(rows)

    df["absolute_difference"] = (
        df["predicted_pct"] -
        df["actual_pct"]
    ).abs()

    df = df.sort_values(
        "actual_count",
        ascending=False
    )

    print(f"\n{split_name.upper()} — DISTRIBUTION COMPARISON")
    print("-" * 110)

    print(
        df.head(top_n).to_string(
            index=False,
            formatters={
                "actual_pct": "{:.4f}%".format,
                "predicted_pct": "{:.4f}%".format,
                "absolute_difference": "{:.4f}".format
            }
        )
    )

    return df


train_distribution_df = distribution_comparison(
    y_train_v8,
    train_pred_ids,
    "Train"
)

val_distribution_df = distribution_comparison(
    y_val_v8,
    val_pred_ids,
    "Validation"
)

test_distribution_df = distribution_comparison(
    y_test_v8,
    test_pred_ids,
    "Test"
)


# =============================================================================
# 31.13 — PREDICTION ENTROPY
# =============================================================================

print("\n" + "=" * 100)
print("31.13 — PREDICTION ENTROPY")
print("=" * 100)

def calculate_entropy(probs):

    # Numerical safety
    probs = np.clip(
        probs,
        1e-12,
        1.0
    )

    entropy = -np.sum(
        probs * np.log(probs),
        axis=1
    )

    return entropy


train_entropy_values = calculate_entropy(train_probs)
val_entropy_values = calculate_entropy(val_probs)
test_entropy_values = calculate_entropy(test_probs)

train_entropy = np.mean(train_entropy_values)
val_entropy = np.mean(val_entropy_values)
test_entropy = np.mean(test_entropy_values)

max_entropy = math.log(VOCAB_SIZE_V8)

print(f"Maximum possible entropy : {max_entropy:.6f}")

print("\nMean prediction entropy:")
print(f"Train      : {train_entropy:.6f}")
print(f"Validation : {val_entropy:.6f}")
print(f"Test       : {test_entropy:.6f}")

print("\nNormalized entropy:")
print(
    f"Train      : "
    f"{train_entropy / max_entropy * 100:.2f}%"
)

print(
    f"Validation : "
    f"{val_entropy / max_entropy * 100:.2f}%"
)

print(
    f"Test       : "
    f"{test_entropy / max_entropy * 100:.2f}%"
)


# =============================================================================
# 31.14 — TOP-1 PROBABILITY ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("31.14 — TOP-1 PROBABILITY ANALYSIS")
print("=" * 100)

def top1_probability_stats(probs):

    top1_probs = np.max(
        probs,
        axis=1
    )

    return {
        "mean": np.mean(top1_probs),
        "median": np.median(top1_probs),
        "min": np.min(top1_probs),
        "max": np.max(top1_probs),
        "p90": np.percentile(top1_probs, 90),
        "p95": np.percentile(top1_probs, 95),
        "p99": np.percentile(top1_probs, 99)
    }


train_top1_stats = top1_probability_stats(train_probs)
val_top1_stats = top1_probability_stats(val_probs)
test_top1_stats = top1_probability_stats(test_probs)

print("\n" + "-" * 90)

for name, stats in [
    ("Train", train_top1_stats),
    ("Validation", val_top1_stats),
    ("Test", test_top1_stats)
]:

    print(f"\n{name}")
    print(f"  Mean   : {stats['mean']:.6f}")
    print(f"  Median : {stats['median']:.6f}")
    print(f"  Min    : {stats['min']:.6f}")
    print(f"  Max    : {stats['max']:.6f}")
    print(f"  P90    : {stats['p90']:.6f}")
    print(f"  P95    : {stats['p95']:.6f}")
    print(f"  P99    : {stats['p99']:.6f}")


# =============================================================================
# 31.15 — CONFIDENCE OF CORRECT VS INCORRECT PREDICTIONS
# =============================================================================

print("\n" + "=" * 100)
print("31.15 — CORRECT VS INCORRECT PREDICTION CONFIDENCE")
print("=" * 100)

def confidence_analysis(probs, targets):

    pred_ids = np.argmax(
        probs,
        axis=1
    )

    top1_probs = np.max(
        probs,
        axis=1
    )

    correct_mask = (
        pred_ids == targets
    )

    correct_probs = top1_probs[correct_mask]
    incorrect_probs = top1_probs[~correct_mask]

    result = {
        "correct_count": int(np.sum(correct_mask)),
        "incorrect_count": int(np.sum(~correct_mask)),
        "accuracy": np.mean(correct_mask),
        "mean_correct_confidence": (
            np.mean(correct_probs)
            if len(correct_probs) > 0
            else np.nan
        ),
        "mean_incorrect_confidence": (
            np.mean(incorrect_probs)
            if len(incorrect_probs) > 0
            else np.nan
        )
    }

    return result


for name, probs, targets in [
    ("Train", train_probs, y_train_v8),
    ("Validation", val_probs, y_val_v8),
    ("Test", test_probs, y_test_v8)
]:

    stats = confidence_analysis(
        probs,
        targets
    )

    print(f"\n{name}")
    print(f"  Correct predictions   : {stats['correct_count']:,}")
    print(f"  Incorrect predictions : {stats['incorrect_count']:,}")
    print(f"  Accuracy              : {stats['accuracy'] * 100:.4f}%")
    print(
        f"  Mean confidence when correct   : "
        f"{stats['mean_correct_confidence']:.6f}"
    )
    print(
        f"  Mean confidence when incorrect : "
        f"{stats['mean_incorrect_confidence']:.6f}"
    )


# =============================================================================
# 31.16 — V7-STYLE COLLAPSE DETECTION
# =============================================================================

print("\n" + "=" * 100)
print("31.16 — PREDICTION COLLAPSE DETECTION")
print("=" * 100)

def collapse_analysis(pred_ids, split_name):

    counter = Counter(pred_ids)

    most_common_id, most_common_count = (
        counter.most_common(1)[0]
    )

    most_common_token = decode_token_id(
        most_common_id
    )

    percentage = (
        most_common_count /
        len(pred_ids) *
        100
    )

    print(f"\n{split_name.upper()}")
    print("-" * 70)

    print(f"Most common prediction : {most_common_token!r}")
    print(f"Token ID               : {most_common_id}")
    print(f"Count                  : {most_common_count:,}")
    print(f"Percentage             : {percentage:.4f}%")
    print(f"Unique predictions     : {len(counter):,}")

    if percentage >= 90:
        status = "🚨 SEVERE COLLAPSE"
    elif percentage >= 70:
        status = "⚠️ STRONG COLLAPSE"
    elif percentage >= 50:
        status = "⚠️ MODERATE COLLAPSE"
    elif percentage >= 30:
        status = "🟡 SOME CONCENTRATION"
    else:
        status = "✅ NO OBVIOUS COLLAPSE"

    print(f"Status                 : {status}")

    return {
        "most_common_id": most_common_id,
        "most_common_token": most_common_token,
        "most_common_count": most_common_count,
        "most_common_percentage": percentage,
        "unique_predictions": len(counter),
        "status": status
    }


train_collapse = collapse_analysis(
    train_pred_ids,
    "Train"
)

val_collapse = collapse_analysis(
    val_pred_ids,
    "Validation"
)

test_collapse = collapse_analysis(
    test_pred_ids,
    "Test"
)


# =============================================================================
# 31.17 — RANDOM BASELINE
# =============================================================================

print("\n" + "=" * 100)
print("31.17 — RANDOM BASELINE")
print("=" * 100)

random_top1 = 1 / VOCAB_SIZE_V8
random_top3 = 3 / VOCAB_SIZE_V8
random_top5 = 5 / VOCAB_SIZE_V8
random_top10 = 10 / VOCAB_SIZE_V8

print(f"Vocabulary size : {VOCAB_SIZE_V8:,}")

print(f"\nRandom Top-1  : {random_top1 * 100:.6f}%")
print(f"Random Top-3  : {random_top3 * 100:.6f}%")
print(f"Random Top-5  : {random_top5 * 100:.6f}%")
print(f"Random Top-10 : {random_top10 * 100:.6f}%")

print("\nV8 TEST PERFORMANCE")
print(f"Top-1  : {results_topk[1]['Test'] * 100:.4f}%")
print(f"Top-3  : {results_topk[3]['Test'] * 100:.4f}%")
print(f"Top-5  : {results_topk[5]['Test'] * 100:.4f}%")
print(f"Top-10 : {results_topk[10]['Test'] * 100:.4f}%")


# =============================================================================
# 31.18 — PERFORMANCE MULTIPLIER OVER RANDOM
# =============================================================================

print("\n" + "=" * 100)
print("31.18 — IMPROVEMENT OVER RANDOM BASELINE")
print("=" * 100)

for k in k_values:

    actual = results_topk[k]["Test"]

    if k == 1:
        random_baseline = random_top1
    elif k == 3:
        random_baseline = random_top3
    elif k == 5:
        random_baseline = random_top5
    else:
        random_baseline = random_top10

    multiplier = actual / random_baseline

    print(
        f"Top-{k:<2} "
        f"→ V8: {actual * 100:.4f}% | "
        f"Random: {random_baseline * 100:.6f}% | "
        f"{multiplier:.2f}× random"
    )


# =============================================================================
# 31.19 — SAMPLE PREDICTIONS
# =============================================================================

print("\n" + "=" * 100)
print("31.19 — SAMPLE TEST PREDICTIONS")
print("=" * 100)

def show_sample_predictions(
    X,
    y,
    probs,
    num_samples=15
):

    num_samples = min(
        num_samples,
        len(X)
    )

    for i in range(num_samples):

        context_ids = X[i]

        context_tokens = [
            decode_token_id(token_id)
            for token_id in context_ids
        ]

        actual_id = int(y[i])
        actual_token = decode_token_id(
            actual_id
        )

        top_ids = np.argsort(
            probs[i]
        )[-5:][::-1]

        print("\n" + "-" * 100)

        print(f"Sample {i + 1}")

        print(
            "Context:"
        )

        print(
            " ".join(context_tokens)
        )

        print(
            f"\nActual next token : "
            f"{actual_token!r} "
            f"(ID {actual_id})"
        )

        print("\nTop-5 predictions:")

        for rank, token_id in enumerate(
            top_ids,
            start=1
        ):

            token = decode_token_id(
                token_id
            )

            probability = probs[
                i,
                token_id
            ]

            print(
                f"  {rank}. "
                f"{token!r:<20} "
                f"ID={token_id:<5} "
                f"P={probability:.6f}"
            )


show_sample_predictions(
    X_test_v8,
    y_test_v8,
    test_probs,
    num_samples=15
)


# =============================================================================
# 31.20 — FINAL FORENSIC SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("📊 STEP 31 — FINAL FORENSIC SUMMARY")
print("=" * 100)

print("\nBEST V8 MODEL")
print("-" * 70)

print("Best checkpoint : Epoch 9")
print(f"Train loss      : {train_loss:.6f}")
print(f"Validation loss : {val_loss:.6f}")
print(f"Test loss       : {test_loss:.6f}")

print("\nACCURACY")
print("-" * 70)

print(f"Train Top-1      : {train_acc * 100:.4f}%")
print(f"Validation Top-1 : {val_acc * 100:.4f}%")
print(f"Test Top-1       : {test_acc * 100:.4f}%")

print(f"Test Top-3       : {results_topk[3]['Test'] * 100:.4f}%")
print(f"Test Top-5       : {results_topk[5]['Test'] * 100:.4f}%")
print(f"Test Top-10      : {results_topk[10]['Test'] * 100:.4f}%")

print("\nPERPLEXITY")
print("-" * 70)

print(f"Train      : {train_perplexity:.4f}")
print(f"Validation : {val_perplexity:.4f}")
print(f"Test       : {test_perplexity:.4f}")

print("\nPREDICTION DIVERSITY")
print("-" * 70)

print(
    f"Train      : "
    f"{train_diversity['unique_count']} unique"
)

print(
    f"Validation : "
    f"{val_diversity['unique_count']} unique"
)

print(
    f"Test       : "
    f"{test_diversity['unique_count']} unique"
)

print("\nENTROPY")
print("-" * 70)

print(f"Train      : {train_entropy:.6f}")
print(f"Validation : {val_entropy:.6f}")
print(f"Test       : {test_entropy:.6f}")
print(f"Maximum    : {max_entropy:.6f}")

print("\nCOLLAPSE CHECK")
print("-" * 70)

print(
    f"Train      : "
    f"{train_collapse['most_common_percentage']:.2f}% → "
    f"{train_collapse['most_common_token']!r}"
)

print(
    f"Validation : "
    f"{val_collapse['most_common_percentage']:.2f}% → "
    f"{val_collapse['most_common_token']!r}"
)

print(
    f"Test       : "
    f"{test_collapse['most_common_percentage']:.2f}% → "
    f"{test_collapse['most_common_token']!r}"
)

print("\n" + "=" * 100)
print("🎯 STEP 31 COMPLETE")
print("=" * 100)

print("""
Next:
    Do NOT change the model yet.

    Send the complete Step 31 output.

    We will use the results to decide whether V8:
        ✅ genuinely learned useful language patterns
        ⚠️ is overfitting
        ⚠️ has prediction concentration
        🚨 has collapsed like V7
        🚀 is ready for text generation

Only after this analysis will we move to Step 32.
""")
```

```python
# =============================================================================
# STEP 32 — V8 TEXT GENERATION
# =============================================================================
# Purpose:
#   Build a reusable text-generation pipeline for the trained V8 LSTM.
#
# Methods:
#   1. Greedy decoding
#   2. Temperature sampling
#   3. Top-K sampling
#   4. Top-P / Nucleus sampling
#
# IMPORTANT:
#   - No retraining
#   - No modification of lstm_v8
#   - Uses the frozen V8 BPE tokenizer
#   - Uses the restored best Epoch-9 model
# =============================================================================

print("=" * 100)
print("📝 STEP 32 — V8 TEXT GENERATION")
print("=" * 100)


# =============================================================================
# 32.1 — RUNTIME CHECK
# =============================================================================

print("\n" + "=" * 100)
print("32.1 — RUNTIME CHECK")
print("=" * 100)

required_variables = [
    "lstm_v8",
    "tokenizer_v8",
    "VOCAB_SIZE_V8",
    "SEQ_LEN_V8",
    "PAD_ID_V8",
    "UNK_ID_V8",
]

missing = [
    var for var in required_variables
    if var not in globals()
]

if missing:
    raise RuntimeError(
        f"❌ Missing required variables: {missing}"
    )

print("✅ All generation dependencies are available.")

print(f"Model       : {lstm_v8.name}")
print(f"Vocabulary  : {VOCAB_SIZE_V8:,}")
print(f"Seq length  : {SEQ_LEN_V8}")
print(f"Tokenizer   : V8 Byte-Level BPE")
print(f"PAD ID      : {PAD_ID_V8}")
print(f"UNK ID      : {UNK_ID_V8}")


# =============================================================================
# 32.2 — TOKENIZER HELPER FUNCTIONS
# =============================================================================

print("\n" + "=" * 100)
print("32.2 — TOKENIZER HELPERS")
print("=" * 100)


def encode_text_v8(text):
    """
    Convert raw text into V8 BPE token IDs.
    """

    if not isinstance(text, str):
        raise TypeError(
            "text must be a string"
        )

    if not text.strip():
        raise ValueError(
            "Input text cannot be empty."
        )

    encoding = tokenizer_v8.encode(
        text
    )

    token_ids = encoding.ids

    if len(token_ids) == 0:
        raise ValueError(
            "Tokenizer produced zero tokens."
        )

    return np.array(
        token_ids,
        dtype=np.int32
    )


def decode_ids_v8(token_ids):
    """
    Decode BPE token IDs back into text.
    """

    token_ids = [
        int(token_id)
        for token_id in token_ids
    ]

    return tokenizer_v8.decode(
        token_ids
    )


def decode_token_v8(token_id):
    """
    Decode a single token ID.
    """

    token = tokenizer_v8.id_to_token(
        int(token_id)
    )

    return token


print("✅ Tokenizer helper functions created.")


# =============================================================================
# 32.3 — BASIC ENCODE / DECODE TEST
# =============================================================================

print("\n" + "=" * 100)
print("32.3 — ENCODE / DECODE TEST")
print("=" * 100)

test_text = "to be or not to be"

test_ids = encode_text_v8(
    test_text
)

test_decoded = decode_ids_v8(
    test_ids
)

print(f"Original text : {test_text!r}")
print(f"Token IDs     : {test_ids.tolist()}")
print(f"Decoded text  : {test_decoded!r}")

assert test_decoded.strip() == test_text

print("\n✅ Encode/decode round-trip passed.")


# =============================================================================
# 32.4 — PREPARE MODEL CONTEXT
# =============================================================================

print("\n" + "=" * 100)
print("32.4 — MODEL CONTEXT PREPARATION")
print("=" * 100)


def prepare_context_v8(token_ids):
    """
    Prepare token IDs for the fixed SEQ_LEN_V8 context.

    If the context is longer than SEQ_LEN_V8:
        keep only the most recent tokens.

    If shorter:
        left-pad with PAD_ID_V8.
    """

    token_ids = list(
        map(int, token_ids)
    )

    if len(token_ids) > SEQ_LEN_V8:

        token_ids = token_ids[
            -SEQ_LEN_V8:
        ]

    elif len(token_ids) < SEQ_LEN_V8:

        padding_length = (
            SEQ_LEN_V8 -
            len(token_ids)
        )

        token_ids = (
            [PAD_ID_V8] * padding_length
            + token_ids
        )

    return np.array(
        [token_ids],
        dtype=np.int32
    )


print("✅ Context preparation function created.")


# =============================================================================
# 32.5 — MODEL PREDICTION FUNCTION
# =============================================================================

print("\n" + "=" * 100)
print("32.5 — NEXT-TOKEN PREDICTION")
print("=" * 100)


def predict_next_token_v8(token_ids):
    """
    Predict the next token probability distribution.
    """

    context = prepare_context_v8(
        token_ids
    )

    probabilities = lstm_v8.predict(
        context,
        verbose=0
    )[0]

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )

    # Numerical safety
    probabilities = np.clip(
        probabilities,
        1e-12,
        1.0
    )

    # Re-normalize
    probabilities /= probabilities.sum()

    return probabilities


print("✅ Next-token prediction function created.")


# =============================================================================
# 32.6 — GREEDY DECODING
# =============================================================================

print("\n" + "=" * 100)
print("32.6 — GREEDY DECODING")
print("=" * 100)


def generate_greedy_v8(
    prompt,
    max_new_tokens=50
):
    """
    Greedy decoding:
        Always choose the highest-probability token.
    """

    token_ids = encode_text_v8(
        prompt
    ).tolist()

    generated_ids = []

    for _ in range(
        max_new_tokens
    ):

        probabilities = predict_next_token_v8(
            token_ids
        )

        # Prevent PAD / UNK generation
        probabilities[PAD_ID_V8] = 0.0
        probabilities[UNK_ID_V8] = 0.0

        next_token_id = int(
            np.argmax(probabilities)
        )

        token_ids.append(
            next_token_id
        )

        generated_ids.append(
            next_token_id
        )

    generated_text = decode_ids_v8(
        token_ids
    )

    return {
        "text": generated_text,
        "generated_ids": generated_ids,
        "all_ids": token_ids
    }


print("✅ Greedy generation function created.")


# =============================================================================
# 32.7 — TEMPERATURE SAMPLING
# =============================================================================

print("\n" + "=" * 100)
print("32.7 — TEMPERATURE SAMPLING")
print("=" * 100)


def apply_temperature_v8(
    probabilities,
    temperature
):
    """
    Apply temperature scaling.

    Lower temperature:
        More deterministic

    Higher temperature:
        More random / diverse
    """

    if temperature <= 0:
        raise ValueError(
            "Temperature must be > 0."
        )

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )

    probabilities = np.clip(
        probabilities,
        1e-12,
        1.0
    )

    logits = np.log(
        probabilities
    )

    scaled_logits = (
        logits / temperature
    )

    scaled_logits -= np.max(
        scaled_logits
    )

    exp_logits = np.exp(
        scaled_logits
    )

    scaled_probabilities = (
        exp_logits /
        np.sum(exp_logits)
    )

    return scaled_probabilities


def sample_temperature_v8(
    probabilities,
    temperature=1.0
):
    """
    Sample one token using temperature.
    """

    scaled_probabilities = (
        apply_temperature_v8(
            probabilities,
            temperature
        )
    )

    # Never generate PAD or UNK
    scaled_probabilities[PAD_ID_V8] = 0.0
    scaled_probabilities[UNK_ID_V8] = 0.0

    scaled_probabilities /= (
        scaled_probabilities.sum()
    )

    token_id = np.random.choice(
        len(scaled_probabilities),
        p=scaled_probabilities
    )

    return int(token_id)


print("✅ Temperature sampling functions created.")


# =============================================================================
# 32.8 — TOP-K SAMPLING
# =============================================================================

print("\n" + "=" * 100)
print("32.8 — TOP-K SAMPLING")
print("=" * 100)


def sample_top_k_v8(
    probabilities,
    k=20,
    temperature=1.0
):
    """
    Top-K sampling.

    Only the K most probable tokens
    are allowed to be sampled.
    """

    if k <= 0:
        raise ValueError(
            "k must be > 0."
        )

    k = min(
        k,
        len(probabilities)
    )

    scaled_probabilities = (
        apply_temperature_v8(
            probabilities,
            temperature
        )
    )

    # Remove PAD and UNK
    scaled_probabilities[
        PAD_ID_V8
    ] = 0.0

    scaled_probabilities[
        UNK_ID_V8
    ] = 0.0

    top_k_ids = np.argpartition(
        scaled_probabilities,
        -k
    )[-k:]

    top_k_probabilities = (
        scaled_probabilities[
            top_k_ids
        ]
    )

    top_k_probabilities /= (
        top_k_probabilities.sum()
    )

    sampled_id = np.random.choice(
        top_k_ids,
        p=top_k_probabilities
    )

    return int(sampled_id)


print("✅ Top-K sampling function created.")


# =============================================================================
# 32.9 — TOP-P / NUCLEUS SAMPLING
# =============================================================================

print("\n" + "=" * 100)
print("32.9 — TOP-P / NUCLEUS SAMPLING")
print("=" * 100)


def sample_top_p_v8(
    probabilities,
    p=0.9,
    temperature=1.0
):
    """
    Top-P / Nucleus sampling.

    Keeps the smallest set of tokens
    whose cumulative probability >= p.
    """

    if not 0 < p <= 1:
        raise ValueError(
            "p must be between 0 and 1."
        )

    scaled_probabilities = (
        apply_temperature_v8(
            probabilities,
            temperature
        )
    )

    # Remove PAD and UNK
    scaled_probabilities[
        PAD_ID_V8
    ] = 0.0

    scaled_probabilities[
        UNK_ID_V8
    ] = 0.0

    sorted_ids = np.argsort(
        scaled_probabilities
    )[::-1]

    sorted_probabilities = (
        scaled_probabilities[
            sorted_ids
        ]
    )

    cumulative_probabilities = np.cumsum(
        sorted_probabilities
    )

    # Tokens after the first point where
    # cumulative probability exceeds p
    remove_mask = (
        cumulative_probabilities > p
    )

    # Always keep at least one token
    remove_mask[0] = False

    sorted_probabilities[
        remove_mask
    ] = 0.0

    sorted_probabilities /= (
        sorted_probabilities.sum()
    )

    sampled_index = np.random.choice(
        len(sorted_ids),
        p=sorted_probabilities
    )

    sampled_id = sorted_ids[
        sampled_index
    ]

    return int(sampled_id)


print("✅ Top-P sampling function created.")


# =============================================================================
# 32.10 — GENERIC SAMPLING GENERATOR
# =============================================================================

print("\n" + "=" * 100)
print("32.10 — GENERIC SAMPLING GENERATOR")
print("=" * 100)


def generate_sampled_v8(
    prompt,
    max_new_tokens=50,
    method="top_k",
    temperature=0.8,
    top_k=20,
    top_p=0.9,
    seed=None
):
    """
    Generate text using sampling.

    method:
        "temperature"
        "top_k"
        "top_p"
    """

    if seed is not None:
        np.random.seed(
            seed
        )

    token_ids = encode_text_v8(
        prompt
    ).tolist()

    generated_ids = []

    for _ in range(
        max_new_tokens
    ):

        probabilities = predict_next_token_v8(
            token_ids
        )

        if method == "temperature":

            next_token_id = (
                sample_temperature_v8(
                    probabilities,
                    temperature
                )
            )

        elif method == "top_k":

            next_token_id = (
                sample_top_k_v8(
                    probabilities,
                    k=top_k,
                    temperature=temperature
                )
            )

        elif method == "top_p":

            next_token_id = (
                sample_top_p_v8(
                    probabilities,
                    p=top_p,
                    temperature=temperature
                )
            )

        else:

            raise ValueError(
                "method must be "
                "'temperature', "
                "'top_k', or 'top_p'."
            )

        token_ids.append(
            next_token_id
        )

        generated_ids.append(
            next_token_id
        )

    generated_text = decode_ids_v8(
        token_ids
    )

    return {
        "text": generated_text,
        "generated_ids": generated_ids,
        "all_ids": token_ids
    }


print("✅ Generic generation function created.")


# =============================================================================
# 32.11 — NEXT TOKEN INSPECTION
# =============================================================================

print("\n" + "=" * 100)
print("32.11 — NEXT TOKEN INSPECTION")
print("=" * 100)


def inspect_next_token_v8(
    prompt,
    top_n=10
):
    """
    Display the model's top-N next-token predictions.
    """

    token_ids = encode_text_v8(
        prompt
    ).tolist()

    probabilities = predict_next_token_v8(
        token_ids
    )

    probabilities[
        PAD_ID_V8
    ] = 0.0

    probabilities[
        UNK_ID_V8
    ] = 0.0

    top_ids = np.argsort(
        probabilities
    )[-top_n:][::-1]

    print(f"\nPrompt: {prompt!r}")
    print("-" * 80)

    for rank, token_id in enumerate(
        top_ids,
        start=1
    ):

        token = decode_token_v8(
            token_id
        )

        probability = probabilities[
            token_id
        ]

        print(
            f"{rank:2d}. "
            f"{token!r:<20} "
            f"ID={token_id:<5} "
            f"P={probability:.6f}"
        )

    return top_ids, probabilities


print("✅ Next-token inspection function created.")


# =============================================================================
# 32.12 — TEST NEXT-TOKEN PREDICTION
# =============================================================================

print("\n" + "=" * 100)
print("32.12 — TESTING NEXT-TOKEN PREDICTION")
print("=" * 100)

inspect_next_token_v8(
    "to be or not to be",
    top_n=10
)

inspect_next_token_v8(
    "what a piece of work is a",
    top_n=10
)

inspect_next_token_v8(
    "the king",
    top_n=10
)


# =============================================================================
# 32.13 — GREEDY GENERATION TESTS
# =============================================================================

print("\n" + "=" * 100)
print("32.13 — GREEDY GENERATION")
print("=" * 100)

greedy_prompts = [
    "to be or not to be",
    "what a piece of work is a",
    "the king",
]

for prompt in greedy_prompts:

    result = generate_greedy_v8(
        prompt,
        max_new_tokens=40
    )

    print("\n" + "-" * 100)
    print(f"PROMPT:\n{prompt}")
    print("-" * 100)
    print("GENERATED:")
    print(result["text"])


# =============================================================================
# 32.14 — TOP-K GENERATION
# =============================================================================

print("\n" + "=" * 100)
print("32.14 — TOP-K SAMPLING")
print("=" * 100)

for prompt in greedy_prompts:

    result = generate_sampled_v8(
        prompt,
        max_new_tokens=40,
        method="top_k",
        temperature=0.8,
        top_k=20,
        seed=42
    )

    print("\n" + "-" * 100)
    print(f"PROMPT:\n{prompt}")
    print("-" * 100)
    print("TOP-K GENERATED:")
    print(result["text"])


# =============================================================================
# 32.15 — TOP-P GENERATION
# =============================================================================

print("\n" + "=" * 100)
print("32.15 — TOP-P / NUCLEUS SAMPLING")
print("=" * 100)

for prompt in greedy_prompts:

    result = generate_sampled_v8(
        prompt,
        max_new_tokens=40,
        method="top_p",
        temperature=0.8,
        top_p=0.9,
        seed=42
    )

    print("\n" + "-" * 100)
    print(f"PROMPT:\n{prompt}")
    print("-" * 100)
    print("TOP-P GENERATED:")
    print(result["text"])


# =============================================================================
# 32.16 — TEMPERATURE COMPARISON
# =============================================================================

print("\n" + "=" * 100)
print("32.16 — TEMPERATURE COMPARISON")
print("=" * 100)

temperature_values = [
    0.5,
    0.7,
    1.0,
    1.2
]

temperature_prompt = (
    "to be or not to be"
)

for temperature in temperature_values:

    result = generate_sampled_v8(
        temperature_prompt,
        max_new_tokens=40,
        method="temperature",
        temperature=temperature,
        seed=42
    )

    print("\n" + "-" * 100)
    print(f"TEMPERATURE = {temperature}")
    print("-" * 100)
    print(result["text"])


# =============================================================================
# 32.17 — GENERATION DIVERSITY TEST
# =============================================================================

print("\n" + "=" * 100)
print("32.17 — GENERATION DIVERSITY TEST")
print("=" * 100)

diversity_prompt = (
    "to be or not to be"
)

generation_results = []

for seed in range(5):

    result = generate_sampled_v8(
        diversity_prompt,
        max_new_tokens=50,
        method="top_p",
        temperature=0.8,
        top_p=0.9,
        seed=seed
    )

    generation_results.append(
        result["text"]
    )

    print("\n" + "-" * 100)
    print(f"SEED = {seed}")
    print("-" * 100)
    print(result["text"])


# =============================================================================
# 32.18 — GENERATION TOKEN DIVERSITY
# =============================================================================

print("\n" + "=" * 100)
print("32.18 — GENERATED TOKEN DIVERSITY")
print("=" * 100)

all_generated_ids = []

for seed in range(10):

    result = generate_sampled_v8(
        diversity_prompt,
        max_new_tokens=50,
        method="top_p",
        temperature=0.8,
        top_p=0.9,
        seed=seed
    )

    all_generated_ids.extend(
        result["generated_ids"]
    )

unique_generated_ids = set(
    all_generated_ids
)

print(
    f"Total generated tokens : "
    f"{len(all_generated_ids):,}"
)

print(
    f"Unique generated tokens: "
    f"{len(unique_generated_ids):,}"
)

print(
    f"Vocabulary utilization  : "
    f"{len(unique_generated_ids) / VOCAB_SIZE_V8 * 100:.2f}%"
)


# =============================================================================
# 32.19 — GENERATION SANITY CHECKS
# =============================================================================

print("\n" + "=" * 100)
print("32.19 — GENERATION SANITY CHECKS")
print("=" * 100)

assert isinstance(
    greedy_prompts[0],
    str
)

assert len(
    greedy_prompts[0]
) > 0

greedy_test = generate_greedy_v8(
    "to be or not to be",
    max_new_tokens=10
)

assert len(
    greedy_test["generated_ids"]
) == 10

assert all(
    token_id != PAD_ID_V8
    for token_id in greedy_test["generated_ids"]
)

assert all(
    token_id != UNK_ID_V8
    for token_id in greedy_test["generated_ids"]
)

top_k_test = generate_sampled_v8(
    "the king",
    max_new_tokens=10,
    method="top_k",
    temperature=0.8,
    top_k=20,
    seed=42
)

assert len(
    top_k_test["generated_ids"]
) == 10

top_p_test = generate_sampled_v8(
    "the king",
    max_new_tokens=10,
    method="top_p",
    temperature=0.8,
    top_p=0.9,
    seed=42
)

assert len(
    top_p_test["generated_ids"]
) == 10

print("✅ Greedy generation works.")
print("✅ Top-K generation works.")
print("✅ Top-P generation works.")
print("✅ No PAD generation.")
print("✅ No UNK generation.")
print("✅ Generated token counts are correct.")


# =============================================================================
# 32.20 — FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("📊 STEP 32 — TEXT GENERATION COMPLETE")
print("=" * 100)

print("""
Generation methods implemented:

    ✅ Greedy decoding
    ✅ Temperature sampling
    ✅ Top-K sampling
    ✅ Top-P / Nucleus sampling

Generation controls:

    Temperature:
        0.5 → conservative
        0.7 → balanced
        1.0 → neutral
        1.2 → more random

    Top-K:
        Controls how many high-probability tokens
        can participate in sampling.

    Top-P:
        Controls the cumulative probability mass
        available for sampling.

IMPORTANT:
    The generated text is now the next thing we need
    to analyze qualitatively.

Do NOT retrain the model yet.

Send the complete Step 32 output.
We will determine:

    1. Is the generated text coherent?
    2. Is it repetitive?
    3. Does it preserve Shakespeare-like structure?
    4. Does punctuation dominate?
    5. Which sampling strategy works best?
    6. Does V8 need architecture improvements?
""")

print("=" * 100)
```

```python
# ============================================================
# STEP 34 — V9 MODEL ARCHITECTURE
# Bidirectional LSTM + Attention
# ============================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

print("=" * 80)
print("STEP 34 — V9 MODEL ARCHITECTURE")
print("Bidirectional LSTM + Attention")
print("=" * 80)


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

tf.keras.utils.set_random_seed(V9_CONFIG["SEED"])

print("\n[1] Random seed")
print(f"    Seed: {V9_CONFIG['SEED']}")
print("    ✓ Random seed configured")


# ============================================================
# 2. READ V9 CONFIGURATION
# ============================================================

VOCAB_SIZE_V9 = V9_CONFIG["VOCAB_SIZE"]
SEQ_LEN_V9 = V9_CONFIG["SEQ_LEN"]
EMBEDDING_DIM_V9 = V9_CONFIG["EMBEDDING_DIM"]
RNN_UNITS_V9 = V9_CONFIG["RNN_UNITS"]
DENSE_UNITS_V9 = V9_CONFIG["DENSE_UNITS"]
DROPOUT_RATE_V9 = V9_CONFIG["DROPOUT_RATE"]
BATCH_SIZE_V9 = V9_CONFIG["BATCH_SIZE"]
LEARNING_RATE_V9 = V9_CONFIG["LEARNING_RATE"]

print("\n[2] Configuration loaded")
print(f"    Vocabulary size : {VOCAB_SIZE_V9}")
print(f"    Sequence length : {SEQ_LEN_V9}")
print(f"    Embedding dim   : {EMBEDDING_DIM_V9}")
print(f"    LSTM units      : {RNN_UNITS_V9}")
print(f"    Dense units     : {DENSE_UNITS_V9}")
print(f"    Dropout         : {DROPOUT_RATE_V9}")
print(f"    Batch size      : {BATCH_SIZE_V9}")
print(f"    Learning rate   : {LEARNING_RATE_V9}")


# ============================================================
# 3. CUSTOM ATTENTION LAYER
# ============================================================

class V9Attention(layers.Layer):
    """
    Attention mechanism for V9.

    Input:
        (batch_size, sequence_length, features)

    Output:
        context vector
        (batch_size, features)

    The layer learns an attention score for every timestep,
    converts scores into normalized attention weights, and
    computes a weighted sum of the timestep representations.
    """

    def __init__(self, **kwargs):
        super().__init__(**kwargs)

        self.score_dense = layers.Dense(
            1,
            activation="tanh",
            name="attention_score"
        )

    def call(self, inputs):
        # ----------------------------------------------------
        # inputs:
        # (batch, seq_len, features)
        # ----------------------------------------------------

        # Calculate attention scores
        scores = self.score_dense(inputs)

        # Shape:
        # (batch, seq_len, 1)

        # Normalize scores across timesteps
        attention_weights = tf.nn.softmax(
            scores,
            axis=1
        )

        # Weighted sum of sequence representations
        context_vector = tf.reduce_sum(
            inputs * attention_weights,
            axis=1
        )

        return context_vector


# ============================================================
# 4. BUILD V9 MODEL
# ============================================================

print("\n[3] Building V9 model...")

inputs_v9 = keras.Input(
    shape=(SEQ_LEN_V9,),
    dtype="int32",
    name="token_input"
)

# ------------------------------------------------------------
# Embedding
# ------------------------------------------------------------

embedding_v9 = layers.Embedding(
    input_dim=VOCAB_SIZE_V9,
    output_dim=EMBEDDING_DIM_V9,
    name="token_embedding"
)(inputs_v9)


# ------------------------------------------------------------
# Bidirectional LSTM
# ------------------------------------------------------------

bilstm_v9 = layers.Bidirectional(
    layers.LSTM(
        RNN_UNITS_V9,
        return_sequences=True,
        name="lstm"
    ),
    name="bidirectional_lstm"
)(embedding_v9)


# ------------------------------------------------------------
# Attention
# ------------------------------------------------------------

attention_v9 = V9Attention(
    name="context_attention"
)(bilstm_v9)


# ------------------------------------------------------------
# Dense projection
# ------------------------------------------------------------

dense_v9 = layers.Dense(
    DENSE_UNITS_V9,
    activation="relu",
    name="projection_dense"
)(attention_v9)


# ------------------------------------------------------------
# Dropout
# ------------------------------------------------------------

dropout_v9 = layers.Dropout(
    DROPOUT_RATE_V9,
    name="projection_dropout"
)(dense_v9)


# ------------------------------------------------------------
# Output layer
# ------------------------------------------------------------

outputs_v9 = layers.Dense(
    VOCAB_SIZE_V9,
    activation="softmax",
    name="next_token_output"
)(dropout_v9)


# ------------------------------------------------------------
# Create model
# ------------------------------------------------------------

lstm_v9 = keras.Model(
    inputs=inputs_v9,
    outputs=outputs_v9,
    name="bilstm_attention_v9"
)


# ============================================================
# 5. MODEL SUMMARY
# ============================================================

print("\n" + "-" * 80)
print("V9 MODEL SUMMARY")
print("-" * 80)

lstm_v9.summary()


# ============================================================
# 6. PARAMETER AUDIT
# ============================================================

print("\n" + "-" * 80)
print("PARAMETER AUDIT")
print("-" * 80)

total_params_v9 = lstm_v9.count_params()

trainable_params_v9 = sum(
    int(tf.size(weight).numpy())
    for weight in lstm_v9.trainable_weights
)

non_trainable_params_v9 = sum(
    int(tf.size(weight).numpy())
    for weight in lstm_v9.non_trainable_weights
)

print(f"Total parameters          : {total_params_v9:,}")
print(f"Trainable parameters      : {trainable_params_v9:,}")
print(f"Non-trainable parameters  : {non_trainable_params_v9:,}")


# ============================================================
# 7. THEORETICAL PARAMETER CALCULATION
# ============================================================

# ------------------------------------------------------------
# Embedding
# ------------------------------------------------------------

embedding_params_v9 = (
    VOCAB_SIZE_V9 *
    EMBEDDING_DIM_V9
)


# ------------------------------------------------------------
# Single LSTM parameter count
#
# LSTM params:
# 4 * [(input_dim * units) +
#      (units * units) +
#      units]
# ------------------------------------------------------------

single_lstm_params_v9 = (
    4 * (
        EMBEDDING_DIM_V9 * RNN_UNITS_V9
        +
        RNN_UNITS_V9 * RNN_UNITS_V9
        +
        RNN_UNITS_V9
    )
)

# Bidirectional = forward + backward
bilstm_params_v9 = (
    single_lstm_params_v9 * 2
)


# ------------------------------------------------------------
# Attention parameters
#
# Dense(features -> 1)
# features = 256 because:
# 128 forward + 128 backward
# ------------------------------------------------------------

bilstm_output_dim_v9 = RNN_UNITS_V9 * 2

attention_params_v9 = (
    bilstm_output_dim_v9 * 1
    + 1
)


# ------------------------------------------------------------
# Dense projection
# ------------------------------------------------------------

projection_params_v9 = (
    bilstm_output_dim_v9 * DENSE_UNITS_V9
    + DENSE_UNITS_V9
)


# ------------------------------------------------------------
# Final vocabulary projection
# ------------------------------------------------------------

output_params_v9 = (
    DENSE_UNITS_V9 * VOCAB_SIZE_V9
    + VOCAB_SIZE_V9
)


theoretical_params_v9 = (
    embedding_params_v9
    + bilstm_params_v9
    + attention_params_v9
    + projection_params_v9
    + output_params_v9
)


print("\nTheoretical parameter calculation:")
print(f"  Embedding             : {embedding_params_v9:,}")
print(f"  Bidirectional LSTM    : {bilstm_params_v9:,}")
print(f"  Attention             : {attention_params_v9:,}")
print(f"  Dense projection      : {projection_params_v9:,}")
print(f"  Output Dense           : {output_params_v9:,}")
print(f"  --------------------------------")
print(f"  Theoretical total      : {theoretical_params_v9:,}")


# ============================================================
# 8. PARAMETER CONSISTENCY CHECK
# ============================================================

print("\nParameter consistency:")

param_check_1 = (
    total_params_v9 ==
    trainable_params_v9 +
    non_trainable_params_v9
)

param_check_2 = (
    total_params_v9 ==
    theoretical_params_v9
)

print(
    f"[{'PASS' if param_check_1 else 'FAIL'}] "
    "Total = trainable + non-trainable"
)

print(
    f"[{'PASS' if param_check_2 else 'FAIL'}] "
    "Keras total = theoretical total"
)


# ============================================================
# 9. EXPECTED TENSOR SHAPES
# ============================================================

print("\n" + "-" * 80)
print("EXPECTED TENSOR SHAPES")
print("-" * 80)

print(f"Input:")
print(f"    (batch, {SEQ_LEN_V9})")

print(f"\nEmbedding:")
print(f"    (batch, {SEQ_LEN_V9}, {EMBEDDING_DIM_V9})")

print(f"\nBidirectional LSTM:")
print(
    f"    (batch, {SEQ_LEN_V9}, "
    f"{RNN_UNITS_V9 * 2})"
)

print("\nAttention:")
print(
    f"    (batch, {RNN_UNITS_V9 * 2})"
)

print("\nDense projection:")
print(
    f"    (batch, {DENSE_UNITS_V9})"
)

print("\nFinal output:")
print(
    f"    (batch, {VOCAB_SIZE_V9})"
)


# ============================================================
# 10. FORWARD PASS TEST
# ============================================================

print("\n" + "-" * 80)
print("FORWARD PASS TEST")
print("-" * 80)

sample_input_v9 = X_train_v8[:1]
sample_target_v9 = y_train_v8[:1]

sample_prediction_v9 = lstm_v9(
    sample_input_v9,
    training=False
)

print(f"Input shape       : {sample_input_v9.shape}")
print(f"Target shape      : {sample_target_v9.shape}")
print(f"Prediction shape  : {sample_prediction_v9.shape}")

forward_shape_check = (
    sample_prediction_v9.shape == (1, VOCAB_SIZE_V9)
)

print(
    f"\n[{'PASS' if forward_shape_check else 'FAIL'}] "
    "Forward-pass output shape"
)


# ============================================================
# 11. SOFTMAX VALIDATION
# ============================================================

prediction_sum_v9 = float(
    tf.reduce_sum(sample_prediction_v9[0]).numpy()
)

prediction_min_v9 = float(
    tf.reduce_min(sample_prediction_v9[0]).numpy()
)

prediction_max_v9 = float(
    tf.reduce_max(sample_prediction_v9[0]).numpy()
)

print("\n" + "-" * 80)
print("SOFTMAX VALIDATION")
print("-" * 80)

print(f"Probability sum : {prediction_sum_v9:.8f}")
print(f"Minimum prob    : {prediction_min_v9:.8f}")
print(f"Maximum prob    : {prediction_max_v9:.8f}")

softmax_check = (
    abs(prediction_sum_v9 - 1.0) < 1e-5
    and prediction_min_v9 >= 0
    and prediction_max_v9 <= 1
)

print(
    f"\n[{'PASS' if softmax_check else 'FAIL'}] "
    "Softmax probability validation"
)


# ============================================================
# 12. ATTENTION WEIGHT VALIDATION
# ============================================================

print("\n" + "-" * 80)
print("ATTENTION MECHANISM VALIDATION")
print("-" * 80)

# Create an intermediate model that exposes BiLSTM output
bilstm_output_model_v9 = keras.Model(
    inputs=lstm_v9.input,
    outputs=lstm_v9.get_layer("bidirectional_lstm").output
)

bilstm_output_sample_v9 = bilstm_output_model_v9(
    sample_input_v9,
    training=False
)

# Calculate attention scores manually
attention_score_layer_v9 = (
    lstm_v9
    .get_layer("context_attention")
    .score_dense
)

attention_scores_v9 = attention_score_layer_v9(
    bilstm_output_sample_v9
)

attention_weights_v9 = tf.nn.softmax(
    attention_scores_v9,
    axis=1
)

attention_weight_sum_v9 = float(
    tf.reduce_sum(attention_weights_v9[0]).numpy()
)

attention_min_v9 = float(
    tf.reduce_min(attention_weights_v9[0]).numpy()
)

attention_max_v9 = float(
    tf.reduce_max(attention_weights_v9[0]).numpy()
)

print(
    f"BiLSTM output shape : "
    f"{bilstm_output_sample_v9.shape}"
)

print(
    f"Attention weights shape : "
    f"{attention_weights_v9.shape}"
)

print(
    f"Attention weight sum : "
    f"{attention_weight_sum_v9:.8f}"
)

print(
    f"Minimum attention weight : "
    f"{attention_min_v9:.8f}"
)

print(
    f"Maximum attention weight : "
    f"{attention_max_v9:.8f}"
)

attention_check = (
    attention_weights_v9.shape == (1, SEQ_LEN_V9, 1)
    and
    abs(attention_weight_sum_v9 - 1.0) < 1e-5
    and
    attention_min_v9 >= 0
)

print(
    f"\n[{'PASS' if attention_check else 'FAIL'}] "
    "Attention weight validation"
)


# ============================================================
# 13. COMPILE MODEL
# ============================================================

print("\n" + "-" * 80)
print("MODEL COMPILATION")
print("-" * 80)

optimizer_v9 = keras.optimizers.Adam(
    learning_rate=LEARNING_RATE_V9
)

loss_v9 = keras.losses.SparseCategoricalCrossentropy()

metric_v9 = keras.metrics.SparseCategoricalAccuracy(
    name="accuracy"
)

lstm_v9.compile(
    optimizer=optimizer_v9,
    loss=loss_v9,
    metrics=[metric_v9]
)

print("Optimizer : Adam")
print(f"Learning rate : {LEARNING_RATE_V9}")
print("Loss      : SparseCategoricalCrossentropy")
print("Metric    : SparseCategoricalAccuracy")
print("\n✓ Model compiled successfully")


# ============================================================
# 14. FINAL ARCHITECTURE CHECK
# ============================================================

print("\n" + "=" * 80)
print("V9 FINAL ARCHITECTURE AUDIT")
print("=" * 80)

final_checks_v9 = {
    "Vocabulary size correct":
        VOCAB_SIZE_V9 == 2500,

    "Sequence length correct":
        SEQ_LEN_V9 == 30,

    "Embedding dimension correct":
        EMBEDDING_DIM_V9 == 128,

    "Bidirectional LSTM present":
        isinstance(
            lstm_v9.get_layer("bidirectional_lstm"),
            layers.Bidirectional
        ),

    "Attention layer present":
        isinstance(
            lstm_v9.get_layer("context_attention"),
            V9Attention
        ),

    "Output vocabulary correct":
        lstm_v9.output_shape[-1] == 2500,

    "Parameter calculation correct":
        param_check_2,

    "Forward pass correct":
        forward_shape_check,

    "Softmax valid":
        softmax_check,

    "Attention weights valid":
        attention_check,

    "Model compiled":
        lstm_v9.optimizer is not None,
}

all_v9_checks_passed = True

for name, result in final_checks_v9.items():

    status = "PASS" if result else "FAIL"

    print(f"[{status}] {name}")

    if not result:
        all_v9_checks_passed = False


print("\n" + "=" * 80)

if all_v9_checks_passed:
    print("STEP 34 STATUS: PASSED")
    print("V9 BiLSTM + Attention architecture is valid.")
    print("NO TRAINING PERFORMED.")
    print("Next: training configuration and final pre-training audit.")
else:
    print("STEP 34 STATUS: FAILED")
    print("Do NOT train V9 until the failed checks are fixed.")

print("=" * 80)
```

```python
# ============================================================
# STEP 37 — V9 FORENSIC EVALUATION
# Bidirectional LSTM + Attention
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf
from collections import Counter

print("=" * 80)
print("STEP 37 — V9 FORENSIC EVALUATION")
print("Bidirectional LSTM + Attention")
print("=" * 80)


# ============================================================
# 1. RUNTIME / MODEL STATUS
# ============================================================

print("\n" + "-" * 80)
print("RUNTIME / MODEL STATUS")
print("-" * 80)

print(f"Model name        : {lstm_v9.name}")
print(f"Model parameters  : {lstm_v9.count_params():,}")
print(f"Vocabulary size   : {VOCAB_SIZE_V9}")
print(f"Sequence length   : {SEQ_LEN_V9}")
print(f"Best epoch        : {best_val_loss_epoch_v9}")
print(f"Best val loss     : {best_val_loss_v9:.6f}")


# ============================================================
# 2. EVALUATION FUNCTION
# ============================================================

def evaluate_split_v9(model, X, y, split_name):
    """
    Evaluate one dataset split.

    Returns:
        loss
        accuracy
    """

    loss, accuracy = model.evaluate(
        X,
        y,
        batch_size=BATCH_SIZE_V9,
        verbose=0
    )

    print(
        f"{split_name:<12} "
        f"Loss = {loss:.6f} | "
        f"Accuracy = {accuracy * 100:.4f}%"
    )

    return float(loss), float(accuracy)


# ============================================================
# 3. BASIC MODEL EVALUATION
# ============================================================

print("\n" + "-" * 80)
print("BASIC MODEL EVALUATION")
print("-" * 80)

train_loss_v9, train_acc_v9 = evaluate_split_v9(
    lstm_v9,
    X_train_v8,
    y_train_v8,
    "Train"
)

val_loss_v9, val_acc_v9 = evaluate_split_v9(
    lstm_v9,
    X_val_v8,
    y_val_v8,
    "Validation"
)

test_loss_v9, test_acc_v9 = evaluate_split_v9(
    lstm_v9,
    X_test_v8,
    y_test_v8,
    "Test"
)


# ============================================================
# 4. PERPLEXITY
# ============================================================

print("\n" + "-" * 80)
print("PERPLEXITY")
print("-" * 80)


def perplexity_from_loss_v9(loss):
    return float(np.exp(loss))


train_ppl_v9 = perplexity_from_loss_v9(train_loss_v9)
val_ppl_v9 = perplexity_from_loss_v9(val_loss_v9)
test_ppl_v9 = perplexity_from_loss_v9(test_loss_v9)

random_ppl_v9 = float(VOCAB_SIZE_V9)


print(f"Train perplexity      : {train_ppl_v9:.4f}")
print(f"Validation perplexity : {val_ppl_v9:.4f}")
print(f"Test perplexity       : {test_ppl_v9:.4f}")
print(f"Random/uniform PPL    : {random_ppl_v9:.4f}")


# ============================================================
# 5. GENERATE PREDICTION PROBABILITIES
# ============================================================

print("\n" + "-" * 80)
print("GENERATING PREDICTION PROBABILITIES")
print("-" * 80)

print("Generating Train predictions...")
train_probs_v9 = lstm_v9.predict(
    X_train_v8,
    batch_size=BATCH_SIZE_V9,
    verbose=1
)

print("Generating Validation predictions...")
val_probs_v9 = lstm_v9.predict(
    X_val_v8,
    batch_size=BATCH_SIZE_V9,
    verbose=1
)

print("Generating Test predictions...")
test_probs_v9 = lstm_v9.predict(
    X_test_v8,
    batch_size=BATCH_SIZE_V9,
    verbose=1
)


# ============================================================
# 6. SHAPE VALIDATION
# ============================================================

print("\n" + "-" * 80)
print("PREDICTION SHAPE VALIDATION")
print("-" * 80)

print(f"Train predictions : {train_probs_v9.shape}")
print(f"Val predictions   : {val_probs_v9.shape}")
print(f"Test predictions  : {test_probs_v9.shape}")

shape_checks_v9 = {
    "Train prediction shape":
        train_probs_v9.shape == (
            len(y_train_v8),
            VOCAB_SIZE_V9
        ),

    "Validation prediction shape":
        val_probs_v9.shape == (
            len(y_val_v8),
            VOCAB_SIZE_V9
        ),

    "Test prediction shape":
        test_probs_v9.shape == (
            len(y_test_v8),
            VOCAB_SIZE_V9
        ),
}

for name, result in shape_checks_v9.items():
    print(
        f"[{'PASS' if result else 'FAIL'}] {name}"
    )


# ============================================================
# 7. PREDICTED TOKEN IDs
# ============================================================

train_pred_ids_v9 = np.argmax(
    train_probs_v9,
    axis=1
)

val_pred_ids_v9 = np.argmax(
    val_probs_v9,
    axis=1
)

test_pred_ids_v9 = np.argmax(
    test_probs_v9,
    axis=1
)


print("\n" + "-" * 80)
print("PREDICTED ID RANGES")
print("-" * 80)

print(
    f"Train: "
    f"{train_pred_ids_v9.min()} → "
    f"{train_pred_ids_v9.max()}"
)

print(
    f"Val  : "
    f"{val_pred_ids_v9.min()} → "
    f"{val_pred_ids_v9.max()}"
)

print(
    f"Test : "
    f"{test_pred_ids_v9.min()} → "
    f"{test_pred_ids_v9.max()}"
)


# ============================================================
# 8. TOP-K ACCURACY
# ============================================================

print("\n" + "-" * 80)
print("TOP-K ACCURACY")
print("-" * 80)


def top_k_accuracy_v9(probs, targets, k):
    top_k_ids = np.argpartition(
        probs,
        -k,
        axis=1
    )[:, -k:]

    correct = np.any(
        top_k_ids == targets[:, None],
        axis=1
    )

    return float(np.mean(correct))


ks_v9 = [1, 3, 5, 10]

topk_results_v9 = {}

for k in ks_v9:

    train_score = top_k_accuracy_v9(
        train_probs_v9,
        y_train_v8,
        k
    )

    val_score = top_k_accuracy_v9(
        val_probs_v9,
        y_val_v8,
        k
    )

    test_score = top_k_accuracy_v9(
        test_probs_v9,
        y_test_v8,
        k
    )

    topk_results_v9[k] = {
        "Train": train_score,
        "Validation": val_score,
        "Test": test_score,
    }

    print(
        f"Top-{k:<2} | "
        f"Train: {train_score * 100:7.4f}% | "
        f"Val: {val_score * 100:7.4f}% | "
        f"Test: {test_score * 100:7.4f}%"
    )


# ============================================================
# 9. PREDICTION DIVERSITY
# ============================================================

print("\n" + "-" * 80)
print("PREDICTION DIVERSITY")
print("-" * 80)


def prediction_diversity_v9(pred_ids):
    unique_ids = np.unique(pred_ids)

    return {
        "unique": len(unique_ids),
        "percentage": (
            len(unique_ids) /
            VOCAB_SIZE_V9 *
            100
        ),
    }


train_diversity_v9 = prediction_diversity_v9(
    train_pred_ids_v9
)

val_diversity_v9 = prediction_diversity_v9(
    val_pred_ids_v9
)

test_diversity_v9 = prediction_diversity_v9(
    test_pred_ids_v9
)


print(
    f"Train: "
    f"{train_diversity_v9['unique']} unique "
    f"({train_diversity_v9['percentage']:.2f}% vocab)"
)

print(
    f"Val:   "
    f"{val_diversity_v9['unique']} unique "
    f"({val_diversity_v9['percentage']:.2f}% vocab)"
)

print(
    f"Test:  "
    f"{test_diversity_v9['unique']} unique "
    f"({test_diversity_v9['percentage']:.2f}% vocab)"
)


# ============================================================
# 10. MOST FREQUENT PREDICTIONS
# ============================================================

print("\n" + "-" * 80)
print("MOST FREQUENT PREDICTED TOKENS")
print("-" * 80)


def token_name_v9(token_id):
    try:
        return tokenizer_v8.id_to_token(int(token_id))
    except Exception:
        try:
            return tokenizer_v8.decode([int(token_id)])
        except Exception:
            return f"<ID:{token_id}>"


def print_top_predictions_v9(pred_ids, split_name, n=20):

    counter = Counter(
        pred_ids.tolist()
    )

    total = len(pred_ids)

    print(f"\n{split_name}")

    print(
        f"{'Rank':<6}"
        f"{'ID':<8}"
        f"{'Token':<20}"
        f"{'Count':<10}"
        f"{'Percent':<10}"
    )

    print("-" * 60)

    for rank, (token_id, count) in enumerate(
        counter.most_common(n),
        start=1
    ):

        token = token_name_v9(token_id)

        percentage = (
            count / total * 100
        )

        print(
            f"{rank:<6}"
            f"{token_id:<8}"
            f"{repr(token):<20}"
            f"{count:<10}"
            f"{percentage:>7.3f}%"
        )


print_top_predictions_v9(
    train_pred_ids_v9,
    "TRAIN",
    n=20
)

print_top_predictions_v9(
    val_pred_ids_v9,
    "VALIDATION",
    n=20
)

print_top_predictions_v9(
    test_pred_ids_v9,
    "TEST",
    n=20
)


# ============================================================
# 11. ACTUAL TEST TARGET DISTRIBUTION
# ============================================================

print("\n" + "-" * 80)
print("ACTUAL TEST TARGET DISTRIBUTION")
print("-" * 80)

actual_test_counter_v9 = Counter(
    y_test_v8.tolist()
)

print(
    f"{'Rank':<6}"
    f"{'ID':<8}"
    f"{'Token':<20}"
    f"{'Count':<10}"
    f"{'Percent':<10}"
)

print("-" * 60)

for rank, (token_id, count) in enumerate(
    actual_test_counter_v9.most_common(20),
    start=1
):

    token = token_name_v9(token_id)

    percentage = (
        count /
        len(y_test_v8) *
        100
    )

    print(
        f"{rank:<6}"
        f"{token_id:<8}"
        f"{repr(token):<20}"
        f"{count:<10}"
        f"{percentage:>7.3f}%"
    )


# ============================================================
# 12. ACTUAL VS PREDICTED TEST DISTRIBUTION
# ============================================================

print("\n" + "-" * 80)
print("ACTUAL VS PREDICTED TEST DISTRIBUTION")
print("-" * 80)


actual_counts_v9 = Counter(
    y_test_v8.tolist()
)

pred_counts_v9 = Counter(
    test_pred_ids_v9.tolist()
)

comparison_ids_v9 = set(
    actual_counts_v9.keys()
) | set(
    pred_counts_v9.keys()
)

distribution_rows_v9 = []

for token_id in comparison_ids_v9:

    actual_count = actual_counts_v9.get(
        token_id,
        0
    )

    predicted_count = pred_counts_v9.get(
        token_id,
        0
    )

    actual_pct = (
        actual_count /
        len(y_test_v8) *
        100
    )

    predicted_pct = (
        predicted_count /
        len(y_test_v8) *
        100
    )

    difference = (
        predicted_pct -
        actual_pct
    )

    distribution_rows_v9.append(
        (
            token_id,
            token_name_v9(token_id),
            actual_count,
            actual_pct,
            predicted_count,
            predicted_pct,
            difference,
        )
    )


distribution_rows_v9.sort(
    key=lambda x: abs(x[6]),
    reverse=True
)


print(
    f"{'ID':<7}"
    f"{'Token':<20}"
    f"{'Actual %':<12}"
    f"{'Predicted %':<14}"
    f"{'Difference':<12}"
)

print("-" * 70)

for row in distribution_rows_v9[:30]:

    print(
        f"{row[0]:<7}"
        f"{repr(row[1]):<20}"
        f"{row[3]:>8.3f}%   "
        f"{row[5]:>9.3f}%     "
        f"{row[6]:>+8.3f} pp"
    )


# ============================================================
# 13. ENTROPY
# ============================================================

print("\n" + "-" * 80)
print("PREDICTION ENTROPY")
print("-" * 80)


def mean_entropy_v9(probs):

    safe_probs = np.clip(
        probs,
        1e-12,
        1.0
    )

    entropy = -np.sum(
        safe_probs *
        np.log(safe_probs),
        axis=1
    )

    return float(np.mean(entropy))


max_entropy_v9 = np.log(
    VOCAB_SIZE_V9
)

train_entropy_v9 = mean_entropy_v9(
    train_probs_v9
)

val_entropy_v9 = mean_entropy_v9(
    val_probs_v9
)

test_entropy_v9 = mean_entropy_v9(
    test_probs_v9
)


print(f"Maximum possible entropy : {max_entropy_v9:.6f}")

print(
    f"Train entropy            : "
    f"{train_entropy_v9:.6f} "
    f"({train_entropy_v9 / max_entropy_v9 * 100:.2f}% max)"
)

print(
    f"Validation entropy       : "
    f"{val_entropy_v9:.6f} "
    f"({val_entropy_v9 / max_entropy_v9 * 100:.2f}% max)"
)

print(
    f"Test entropy             : "
    f"{test_entropy_v9:.6f} "
    f"({test_entropy_v9 / max_entropy_v9 * 100:.2f}% max)"
)


# ============================================================
# 14. TOP-1 CONFIDENCE
# ============================================================

print("\n" + "-" * 80)
print("TOP-1 CONFIDENCE")
print("-" * 80)


def confidence_statistics_v9(probs, targets):

    pred_ids = np.argmax(
        probs,
        axis=1
    )

    top1_probs = np.max(
        probs,
        axis=1
    )

    correct_mask = (
        pred_ids == targets
    )

    return {
        "mean": float(np.mean(top1_probs)),
        "median": float(np.median(top1_probs)),
        "min": float(np.min(top1_probs)),
        "max": float(np.max(top1_probs)),
        "p90": float(np.percentile(top1_probs, 90)),
        "p95": float(np.percentile(top1_probs, 95)),
        "p99": float(np.percentile(top1_probs, 99)),
        "correct_count": int(np.sum(correct_mask)),
        "incorrect_count": int(np.sum(~correct_mask)),
        "correct_mean": (
            float(np.mean(top1_probs[correct_mask]))
            if np.any(correct_mask)
            else np.nan
        ),
        "incorrect_mean": (
            float(np.mean(top1_probs[~correct_mask]))
            if np.any(~correct_mask)
            else np.nan
        ),
    }


confidence_train_v9 = confidence_statistics_v9(
    train_probs_v9,
    y_train_v8
)

confidence_val_v9 = confidence_statistics_v9(
    val_probs_v9,
    y_val_v8
)

confidence_test_v9 = confidence_statistics_v9(
    test_probs_v9,
    y_test_v8
)


for name, stats in [
    ("Train", confidence_train_v9),
    ("Validation", confidence_val_v9),
    ("Test", confidence_test_v9),
]:

    print(f"\n{name}")

    print(
        f"  Mean     : {stats['mean']:.6f}"
    )

    print(
        f"  Median   : {stats['median']:.6f}"
    )

    print(
        f"  Min      : {stats['min']:.6f}"
    )

    print(
        f"  Max      : {stats['max']:.6f}"
    )

    print(
        f"  P90      : {stats['p90']:.6f}"
    )

    print(
        f"  P95      : {stats['p95']:.6f}"
    )

    print(
        f"  P99      : {stats['p99']:.6f}"
    )

    print(
        f"  Correct mean confidence   : "
        f"{stats['correct_mean']:.6f}"
    )

    print(
        f"  Incorrect mean confidence : "
        f"{stats['incorrect_mean']:.6f}"
    )


# ============================================================
# 15. RANDOM BASELINE
# ============================================================

print("\n" + "-" * 80)
print("RANDOM BASELINE")
print("-" * 80)

random_top1_v9 = 1 / VOCAB_SIZE_V9
random_top3_v9 = 3 / VOCAB_SIZE_V9
random_top5_v9 = 5 / VOCAB_SIZE_V9
random_top10_v9 = 10 / VOCAB_SIZE_V9

print(
    f"Random Top-1  : {random_top1_v9 * 100:.4f}%"
)

print(
    f"Random Top-3  : {random_top3_v9 * 100:.4f}%"
)

print(
    f"Random Top-5  : {random_top5_v9 * 100:.4f}%"
)

print(
    f"Random Top-10 : {random_top10_v9 * 100:.4f}%"
)


print("\nV9 Test improvement over random:")

for k in ks_v9:

    v9_score = topk_results_v9[k]["Test"]

    random_score = k / VOCAB_SIZE_V9

    multiplier = (
        v9_score /
        random_score
    )

    print(
        f"Top-{k:<2}: "
        f"{multiplier:.2f}x random"
    )


# ============================================================
# 16. COLLAPSE ANALYSIS
# ============================================================

print("\n" + "-" * 80)
print("PREDICTION COLLAPSE ANALYSIS")
print("-" * 80)


def collapse_analysis_v9(pred_ids, split_name):

    counts = Counter(
        pred_ids.tolist()
    )

    total = len(pred_ids)

    most_common_id, most_common_count = (
        counts.most_common(1)[0]
    )

    concentration = (
        most_common_count /
        total *
        100
    )

    unique_count = len(
        counts
    )

    print(f"\n{split_name}")

    print(
        f"Most common prediction : "
        f"{repr(token_name_v9(most_common_id))}"
    )

    print(
        f"Count                  : "
        f"{most_common_count:,}"
    )

    print(
        f"Concentration          : "
        f"{concentration:.4f}%"
    )

    print(
        f"Unique predictions     : "
        f"{unique_count}"
    )

    if concentration >= 50:
        status = "SEVERE COLLAPSE"
    elif concentration >= 30:
        status = "HIGH CONCENTRATION"
    elif concentration >= 20:
        status = "MODERATE CONCENTRATION"
    else:
        status = "NO OBVIOUS COLLAPSE"

    print(f"Status                  : {status}")

    return {
        "token_id": most_common_id,
        "count": most_common_count,
        "concentration": concentration,
        "unique": unique_count,
        "status": status,
    }


collapse_train_v9 = collapse_analysis_v9(
    train_pred_ids_v9,
    "TRAIN"
)

collapse_val_v9 = collapse_analysis_v9(
    val_pred_ids_v9,
    "VALIDATION"
)

collapse_test_v9 = collapse_analysis_v9(
    test_pred_ids_v9,
    "TEST"
)


# ============================================================
# 17. CORRECT / INCORRECT TEST PREDICTIONS
# ============================================================

print("\n" + "-" * 80)
print("TEST CORRECT / INCORRECT PREDICTIONS")
print("-" * 80)

test_correct_mask_v9 = (
    test_pred_ids_v9 ==
    y_test_v8
)

correct_test_count_v9 = int(
    np.sum(test_correct_mask_v9)
)

incorrect_test_count_v9 = int(
    np.sum(~test_correct_mask_v9)
)

print(
    f"Correct predictions   : "
    f"{correct_test_count_v9:,}"
)

print(
    f"Incorrect predictions : "
    f"{incorrect_test_count_v9:,}"
)

print(
    f"Accuracy              : "
    f"{correct_test_count_v9 / len(y_test_v8) * 100:.4f}%"
)


# ============================================================
# 18. SAMPLE TEST PREDICTIONS
# ============================================================

print("\n" + "-" * 80)
print("SAMPLE TEST PREDICTIONS")
print("-" * 80)


sample_indices_v9 = np.linspace(
    0,
    len(X_test_v8) - 1,
    15,
    dtype=int
)


for display_number, idx in enumerate(
    sample_indices_v9,
    start=1
):

    actual_id = int(
        y_test_v8[idx]
    )

    predicted_id = int(
        test_pred_ids_v9[idx]
    )

    actual_token = token_name_v9(
        actual_id
    )

    predicted_token = token_name_v9(
        predicted_id
    )

    top10_ids = np.argsort(
        test_probs_v9[idx]
    )[-10:][::-1]

    print(
        f"\nSample {display_number:02d} "
        f"(index {idx})"
    )

    print(
        f"Actual target    : "
        f"{repr(actual_token)} "
        f"(ID {actual_id})"
    )

    print(
        f"Predicted target : "
        f"{repr(predicted_token)} "
        f"(ID {predicted_id})"
    )

    print(
        "Top-10 predictions:"
    )

    for rank, token_id in enumerate(
        top10_ids,
        start=1
    ):

        probability = (
            test_probs_v9[
                idx,
                token_id
            ]
        )

        print(
            f"  {rank:02d}. "
            f"{repr(token_name_v9(token_id)):<20} "
            f"{probability:.6f}"
        )


# ============================================================
# 19. ATTENTION ANALYSIS
# ============================================================

print("\n" + "-" * 80)
print("ATTENTION ANALYSIS")
print("-" * 80)


# ------------------------------------------------------------
# Build intermediate model
# ------------------------------------------------------------

bilstm_output_model_v9 = keras.Model(
    inputs=lstm_v9.input,
    outputs=lstm_v9.get_layer(
        "bidirectional_lstm"
    ).output
)


bilstm_test_output_v9 = (
    bilstm_output_model_v9.predict(
        X_test_v8,
        batch_size=BATCH_SIZE_V9,
        verbose=1
    )
)


# ------------------------------------------------------------
# Attention score layer
# ------------------------------------------------------------

attention_layer_v9 = (
    lstm_v9.get_layer(
        "context_attention"
    )
)

attention_score_layer_v9 = (
    attention_layer_v9.score_dense
)


attention_scores_test_v9 = (
    attention_score_layer_v9(
        bilstm_test_output_v9
    )
)


attention_weights_test_v9 = (
    tf.nn.softmax(
        attention_scores_test_v9,
        axis=1
    ).numpy()
)


print(
    f"BiLSTM representations : "
    f"{bilstm_test_output_v9.shape}"
)

print(
    f"Attention weights      : "
    f"{attention_weights_test_v9.shape}"
)


# ------------------------------------------------------------
# Attention statistics
# ------------------------------------------------------------

attention_sums_v9 = (
    np.sum(
        attention_weights_test_v9,
        axis=1
    )
)

attention_max_per_sample_v9 = (
    np.max(
        attention_weights_test_v9,
        axis=1
    )
)

attention_min_per_sample_v9 = (
    np.min(
        attention_weights_test_v9,
        axis=1
    )
)


print(
    f"\nAttention sum mean : "
    f"{np.mean(attention_sums_v9):.8f}"
)

print(
    f"Attention max mean : "
    f"{np.mean(attention_max_per_sample_v9):.6f}"
)

print(
    f"Attention min mean : "
    f"{np.mean(attention_min_per_sample_v9):.6f}"
)


# ------------------------------------------------------------
# Uniform attention reference
# ------------------------------------------------------------

uniform_attention_v9 = (
    1 / SEQ_LEN_V9
)

print(
    f"\nUniform attention weight: "
    f"{uniform_attention_v9:.6f}"
)


# ------------------------------------------------------------
# Average attention by timestep
# ------------------------------------------------------------

mean_attention_by_position_v9 = (
    np.mean(
        attention_weights_test_v9[:, :, 0],
        axis=0
    )
)


print("\nMean attention by timestep:")

for position, weight in enumerate(
    mean_attention_by_position_v9
):

    print(
        f"  Position {position + 1:02d}: "
        f"{weight:.6f}"
    )


# ============================================================
# 20. V8 VS V9 COMPARISON
# ============================================================

print("\n" + "=" * 80)
print("V8 VS V9 COMPARISON")
print("=" * 80)


comparison_metrics_v9 = {

    "Test Accuracy":
        (
            V8_BASELINE["test_accuracy"],
            test_acc_v9
        ),

    "Test Top-3":
        (
            V8_BASELINE["test_top3"],
            topk_results_v9[3]["Test"]
        ),

    "Test Top-5":
        (
            V8_BASELINE["test_top5"],
            topk_results_v9[5]["Test"]
        ),

    "Test Top-10":
        (
            V8_BASELINE["test_top10"],
            topk_results_v9[10]["Test"]
        ),

    "Test Perplexity":
        (
            V8_BASELINE["test_perplexity"],
            test_ppl_v9
        ),
}


print(
    f"{'Metric':<25}"
    f"{'V8':<15}"
    f"{'V9':<15}"
    f"{'Change':<15}"
)

print("-" * 70)


for metric, (v8_value, v9_value) in (
    comparison_metrics_v9.items()
):

    if "Perplexity" in metric:

        change = v9_value - v8_value

        print(
            f"{metric:<25}"
            f"{v8_value:<15.4f}"
            f"{v9_value:<15.4f}"
            f"{change:+.4f}"
        )

    else:

        change = (
            v9_value -
            v8_value
        ) * 100

        print(
            f"{metric:<25}"
            f"{v8_value * 100:<14.4f}%"
            f"{v9_value * 100:<14.4f}%"
            f"{change:+.4f} pp"
        )


# ============================================================
# 21. V9 IMPROVEMENT / REGRESSION SUMMARY
# ============================================================

print("\n" + "-" * 80)
print("V9 IMPROVEMENT / REGRESSION SUMMARY")
print("-" * 80)


accuracy_change_v9 = (
    test_acc_v9 -
    V8_BASELINE["test_accuracy"]
)

top3_change_v9 = (
    topk_results_v9[3]["Test"] -
    V8_BASELINE["test_top3"]
)

top5_change_v9 = (
    topk_results_v9[5]["Test"] -
    V8_BASELINE["test_top5"]
)

top10_change_v9 = (
    topk_results_v9[10]["Test"] -
    V8_BASELINE["test_top10"]
)

ppl_change_v9 = (
    test_ppl_v9 -
    V8_BASELINE["test_perplexity"]
)


def print_change_v9(name, change, lower_is_better=False):

    if lower_is_better:

        status = (
            "IMPROVED"
            if change < 0
            else "REGRESSED"
            if change > 0
            else "UNCHANGED"
        )

        print(
            f"{name:<20}: "
            f"{change:+.4f} "
            f"→ {status}"
        )

    else:

        status = (
            "IMPROVED"
            if change > 0
            else "REGRESSED"
            if change < 0
            else "UNCHANGED"
        )

        print(
            f"{name:<20}: "
            f"{change * 100:+.4f} pp "
            f"→ {status}"
        )


print_change_v9(
    "Top-1 accuracy",
    accuracy_change_v9
)

print_change_v9(
    "Top-3 accuracy",
    top3_change_v9
)

print_change_v9(
    "Top-5 accuracy",
    top5_change_v9
)

print_change_v9(
    "Top-10 accuracy",
    top10_change_v9
)

print_change_v9(
    "Perplexity",
    ppl_change_v9,
    lower_is_better=True
)


# ============================================================
# 22. FINAL FORENSIC STATUS
# ============================================================

print("\n" + "=" * 80)
print("STEP 37 — FINAL FORENSIC STATUS")
print("=" * 80)


all_shape_checks_pass_v9 = all(
    shape_checks_v9.values()
)

finite_prediction_check_v9 = (
    np.all(np.isfinite(train_probs_v9))
    and
    np.all(np.isfinite(val_probs_v9))
    and
    np.all(np.isfinite(test_probs_v9))
)

probability_sum_check_v9 = (
    np.allclose(
        np.sum(train_probs_v9, axis=1),
        1.0,
        atol=1e-5
    )
    and
    np.allclose(
        np.sum(val_probs_v9, axis=1),
        1.0,
        atol=1e-5
    )
    and
    np.allclose(
        np.sum(test_probs_v9, axis=1),
        1.0,
        atol=1e-5
    )
)

attention_sum_check_v9 = (
    np.allclose(
        attention_sums_v9,
        1.0,
        atol=1e-5
    )
)


final_forensic_checks_v9 = {

    "Prediction shapes valid":
        all_shape_checks_pass_v9,

    "Predictions finite":
        finite_prediction_check_v9,

    "Probability sums valid":
        probability_sum_check_v9,

    "Attention sums valid":
        attention_sum_check_v9,

    "Test evaluation completed":
        True,
}


for name, result in final_forensic_checks_v9.items():

    print(
        f"[{'PASS' if result else 'FAIL'}] "
        f"{name}"
    )


if all(final_forensic_checks_v9.values()):

    print("\n" + "=" * 80)
    print("STEP 37 STATUS: PASSED")
    print("=" * 80)

    print("""
V9 forensic evaluation completed successfully.

The test set has now been evaluated.

Next step:
    STEP 38 — V8 vs V9 Analysis + Generation Evaluation
""")

else:

    print("\n" + "=" * 80)
    print("STEP 37 STATUS: FAILED")
    print("=" * 80)

    print(
        "One or more forensic checks failed."
    )

print("=" * 80)
```

```python
# ============================================================
# STEP 38.1 — V10 CAUSAL TRANSFORMER CONFIGURATION
# ============================================================

import os
import random
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# ------------------------------------------------------------
# Reproducibility
# ------------------------------------------------------------

SEED_V10 = 42

os.environ["PYTHONHASHSEED"] = str(SEED_V10)

random.seed(SEED_V10)
np.random.seed(SEED_V10)
tf.random.set_seed(SEED_V10)

# ------------------------------------------------------------
# Frozen V8 tokenizer / data settings
# ------------------------------------------------------------

VOCAB_SIZE_V10 = VOCAB_SIZE_V8
SEQ_LEN_V10 = SEQ_LEN_V8

# ------------------------------------------------------------
# Transformer configuration
# ------------------------------------------------------------

EMBEDDING_DIM_V10 = 128

NUM_TRANSFORMER_BLOCKS_V10 = 2

NUM_HEADS_V10 = 4
KEY_DIM_V10 = 32

FF_DIM_V10 = 256

DROPOUT_RATE_V10 = 0.30

LEARNING_RATE_V10 = 0.001

BATCH_SIZE_V10 = 64

MAX_EPOCHS_V10 = 30

# ------------------------------------------------------------
# Print configuration
# ------------------------------------------------------------

print("=" * 70)
print("V10 CAUSAL TRANSFORMER — FROZEN CONFIGURATION")
print("=" * 70)

print(f"Vocabulary Size       : {VOCAB_SIZE_V10}")
print(f"Sequence Length       : {SEQ_LEN_V10}")
print(f"Embedding Dimension   : {EMBEDDING_DIM_V10}")
print(f"Transformer Blocks    : {NUM_TRANSFORMER_BLOCKS_V10}")
print(f"Attention Heads       : {NUM_HEADS_V10}")
print(f"Key Dimension         : {KEY_DIM_V10}")
print(f"Feed Forward Dimension: {FF_DIM_V10}")
print(f"Dropout               : {DROPOUT_RATE_V10}")
print(f"Learning Rate         : {LEARNING_RATE_V10}")
print(f"Batch Size            : {BATCH_SIZE_V10}")
print(f"Maximum Epochs        : {MAX_EPOCHS_V10}")

print("=" * 70)
```

```python
# ============================================================
# STEP 38.2 — CAUSAL TRANSFORMER BLOCK
# ============================================================

class CausalTransformerBlock(layers.Layer):

    def __init__(
        self,
        embedding_dim,
        num_heads,
        key_dim,
        ff_dim,
        dropout_rate=0.1,
        **kwargs
    ):
        super().__init__(**kwargs)

        self.embedding_dim = embedding_dim
        self.num_heads = num_heads
        self.key_dim = key_dim
        self.ff_dim = ff_dim
        self.dropout_rate = dropout_rate

        # ----------------------------------------------------
        # Multi-Head Self-Attention
        #
        # use_causal_mask=True ensures that token t cannot
        # attend to future tokens.
        # ----------------------------------------------------

        self.attention = layers.MultiHeadAttention(
            num_heads=num_heads,
            key_dim=key_dim,
            dropout=dropout_rate,
            name="causal_self_attention"
        )

        # ----------------------------------------------------
        # Feed Forward Network
        # ----------------------------------------------------

        self.ffn = keras.Sequential(
            [
                layers.Dense(
                    ff_dim,
                    activation="gelu",
                    name="ffn_dense_1"
                ),

                layers.Dropout(
                    dropout_rate,
                    name="ffn_dropout"
                ),

                layers.Dense(
                    embedding_dim,
                    name="ffn_dense_2"
                )
            ],
            name="feed_forward_network"
        )

        # ----------------------------------------------------
        # Layer Normalization
        # ----------------------------------------------------

        self.layernorm1 = layers.LayerNormalization(
            epsilon=1e-6,
            name="layer_norm_1"
        )

        self.layernorm2 = layers.LayerNormalization(
            epsilon=1e-6,
            name="layer_norm_2"
        )

        # ----------------------------------------------------
        # Residual dropout
        # ----------------------------------------------------

        self.dropout1 = layers.Dropout(
            dropout_rate,
            name="attention_dropout"
        )

        self.dropout2 = layers.Dropout(
            dropout_rate,
            name="residual_dropout"
        )

    def call(self, x, training=False):

        # ----------------------------------------------------
        # Self-attention
        #
        # causal=True prevents looking into the future.
        # ----------------------------------------------------

        attention_output = self.attention(
            query=x,
            value=x,
            key=x,
            use_causal_mask=True,
            training=training
        )

        attention_output = self.dropout1(
            attention_output,
            training=training
        )

        # ----------------------------------------------------
        # Residual connection + LayerNorm
        # ----------------------------------------------------

        x = self.layernorm1(
            x + attention_output
        )

        # ----------------------------------------------------
        # Feed Forward Network
        # ----------------------------------------------------

        ffn_output = self.ffn(
            x,
            training=training
        )

        ffn_output = self.dropout2(
            ffn_output,
            training=training
        )

        # ----------------------------------------------------
        # Residual connection + LayerNorm
        # ----------------------------------------------------

        x = self.layernorm2(
            x + ffn_output
        )

        return x

    def get_config(self):

        config = super().get_config()

        config.update(
            {
                "embedding_dim": self.embedding_dim,
                "num_heads": self.num_heads,
                "key_dim": self.key_dim,
                "ff_dim": self.ff_dim,
                "dropout_rate": self.dropout_rate
            }
        )

        return config


print("CausalTransformerBlock created successfully.")
```

```python
# ============================================================
# STEP 38.3 — TOKEN + POSITION EMBEDDING
# ============================================================

class TokenAndPositionEmbedding(layers.Layer):

    def __init__(
        self,
        vocab_size,
        sequence_length,
        embedding_dim,
        **kwargs
    ):
        super().__init__(**kwargs)

        self.vocab_size = vocab_size
        self.sequence_length = sequence_length
        self.embedding_dim = embedding_dim

        # Token embedding
        self.token_embedding = layers.Embedding(
            input_dim=vocab_size,
            output_dim=embedding_dim,
            name="token_embedding"
        )

        # Learned positional embedding
        self.position_embedding = layers.Embedding(
            input_dim=sequence_length,
            output_dim=embedding_dim,
            name="position_embedding"
        )

    def call(self, inputs):

        # Sequence positions:
        # [0, 1, 2, ..., sequence_length-1]

        positions = tf.range(
            start=0,
            limit=tf.shape(inputs)[-1],
            delta=1
        )

        # Position embeddings
        position_embeddings = self.position_embedding(
            positions
        )

        # Token embeddings
        token_embeddings = self.token_embedding(
            inputs
        )

        # Broadcasting adds the same positional vector
        # to the corresponding position in every sequence.

        return token_embeddings + position_embeddings

    def get_config(self):

        config = super().get_config()

        config.update(
            {
                "vocab_size": self.vocab_size,
                "sequence_length": self.sequence_length,
                "embedding_dim": self.embedding_dim
            }
        )

        return config


print("TokenAndPositionEmbedding created successfully.")
```

```python
# ============================================================
# STEP 38.4 — BUILD V10 CAUSAL TRANSFORMER
# ============================================================

def build_v10_transformer():

    # --------------------------------------------------------
    # Input
    # --------------------------------------------------------

    inputs = keras.Input(
        shape=(SEQ_LEN_V10,),
        dtype="int32",
        name="token_ids"
    )

    # --------------------------------------------------------
    # Token + positional embeddings
    # --------------------------------------------------------

    x = TokenAndPositionEmbedding(
        vocab_size=VOCAB_SIZE_V10,
        sequence_length=SEQ_LEN_V10,
        embedding_dim=EMBEDDING_DIM_V10,
        name="token_position_embedding"
    )(inputs)

    # --------------------------------------------------------
    # Transformer blocks
    # --------------------------------------------------------

    for block_idx in range(NUM_TRANSFORMER_BLOCKS_V10):

        x = CausalTransformerBlock(
            embedding_dim=EMBEDDING_DIM_V10,
            num_heads=NUM_HEADS_V10,
            key_dim=KEY_DIM_V10,
            ff_dim=FF_DIM_V10,
            dropout_rate=DROPOUT_RATE_V10,
            name=f"transformer_block_{block_idx + 1}"
        )(x)

    # --------------------------------------------------------
    # Final dropout
    # --------------------------------------------------------

    x = layers.Dropout(
        DROPOUT_RATE_V10,
        name="final_dropout"
    )(x)

    # --------------------------------------------------------
    # Vocabulary projection
    #
    # One probability distribution for every input position.
    #
    # Shape:
    # (batch, 30, 2500)
    # --------------------------------------------------------

    outputs = layers.Dense(
        VOCAB_SIZE_V10,
        activation="softmax",
        name="next_token_prediction"
    )(x)

    model = keras.Model(
        inputs=inputs,
        outputs=outputs,
        name="causal_transformer_v10"
    )

    return model


transformer_v10 = build_v10_transformer()

print("\nV10 model created successfully.")
```

```python
# ============================================================
# STEP 38.8 — CAUSAL MASK CONCEPTUAL AUDIT
# ============================================================

def create_causal_mask(seq_len):

    """
    Creates a lower-triangular causal attention mask.

    True  = attention is allowed
    False = attention is blocked
    """

    return tf.linalg.band_part(
        tf.ones(
            (seq_len, seq_len),
            dtype=tf.bool
        ),
        -1,
        0
    )


causal_mask_v10 = create_causal_mask(
    SEQ_LEN_V10
)

print("=" * 70)
print("V10 CAUSAL MASK")
print("=" * 70)

print(
    causal_mask_v10.numpy().astype(int)
)

# ------------------------------------------------------------
# Validate lower triangular structure
# ------------------------------------------------------------

mask_np = causal_mask_v10.numpy()

for i in range(SEQ_LEN_V10):
    for j in range(SEQ_LEN_V10):

        if j > i:
            assert mask_np[i, j] == False

        else:
            assert mask_np[i, j] == True

print("=" * 70)
print("✓ Causal mask structure PASSED")
print("✓ Future-token access is blocked")
print("=" * 70)
```

```python
# ============================================================
# STEP 39.2 — CORRECTED V10 CAUSAL TRANSFORMER
# ============================================================

def build_v10_transformer():

    # --------------------------------------------------------
    # Input
    # --------------------------------------------------------

    inputs = keras.Input(
        shape=(SEQ_LEN_V10,),
        dtype="int32",
        name="token_ids"
    )

    # --------------------------------------------------------
    # Token + positional embeddings
    # --------------------------------------------------------

    x = TokenAndPositionEmbedding(
        vocab_size=VOCAB_SIZE_V10,
        sequence_length=SEQ_LEN_V10,
        embedding_dim=EMBEDDING_DIM_V10,
        name="token_position_embedding"
    )(inputs)

    # --------------------------------------------------------
    # Transformer blocks
    # --------------------------------------------------------

    for block_idx in range(NUM_TRANSFORMER_BLOCKS_V10):

        x = CausalTransformerBlock(
            embedding_dim=EMBEDDING_DIM_V10,
            num_heads=NUM_HEADS_V10,
            key_dim=KEY_DIM_V10,
            ff_dim=FF_DIM_V10,
            dropout_rate=DROPOUT_RATE_V10,
            name=f"transformer_block_{block_idx + 1}"
        )(x)

    # --------------------------------------------------------
    # Final dropout
    # --------------------------------------------------------

    x = layers.Dropout(
        DROPOUT_RATE_V10,
        name="final_dropout"
    )(x)

    # --------------------------------------------------------
    # IMPORTANT:
    #
    # Take ONLY the final sequence position.
    #
    # Shape:
    # (batch, 30, 128)
    #
    #             ↓
    #
    # (batch, 128)
    # --------------------------------------------------------

    x = layers.Lambda(
        lambda tensor: tensor[:, -1, :],
        name="last_position"
    )(x)

    # --------------------------------------------------------
    # Next-token prediction
    #
    # Shape:
    # (batch, 2500)
    # --------------------------------------------------------

    outputs = layers.Dense(
        VOCAB_SIZE_V10,
        activation="softmax",
        name="next_token_prediction"
    )(x)

    model = keras.Model(
        inputs=inputs,
        outputs=outputs,
        name="causal_transformer_v10"
    )

    return model


# ------------------------------------------------------------
# Rebuild model
# ------------------------------------------------------------

transformer_v10 = build_v10_transformer()

print("=" * 80)
print("CORRECTED V10 MODEL CREATED")
print("=" * 80)
```

```python
# ============================================================
# STEP 41.5 — V10 TOP-K ACCURACY
# ============================================================

def calculate_top_k_accuracy(
    predictions,
    targets,
    k
):

    top_k_ids = np.argpartition(
        predictions,
        -k,
        axis=1
    )[:, -k:]

    correct = np.any(
        top_k_ids == targets[:, None],
        axis=1
    )

    return np.mean(correct)


v10_train_top1 = calculate_top_k_accuracy(
    train_predictions_v10,
    y_train_v10,
    1
)

v10_train_top3 = calculate_top_k_accuracy(
    train_predictions_v10,
    y_train_v10,
    3
)

v10_train_top5 = calculate_top_k_accuracy(
    train_predictions_v10,
    y_train_v10,
    5
)

v10_train_top10 = calculate_top_k_accuracy(
    train_predictions_v10,
    y_train_v10,
    10
)

v10_val_top1 = calculate_top_k_accuracy(
    val_predictions_v10,
    y_val_v10,
    1
)

v10_val_top3 = calculate_top_k_accuracy(
    val_predictions_v10,
    y_val_v10,
    3
)

v10_val_top5 = calculate_top_k_accuracy(
    val_predictions_v10,
    y_val_v10,
    5
)

v10_val_top10 = calculate_top_k_accuracy(
    val_predictions_v10,
    y_val_v10,
    10
)

v10_test_top1 = calculate_top_k_accuracy(
    test_predictions_v10,
    y_test_v10,
    1
)

v10_test_top3 = calculate_top_k_accuracy(
    test_predictions_v10,
    y_test_v10,
    3
)

v10_test_top5 = calculate_top_k_accuracy(
    test_predictions_v10,
    y_test_v10,
    5
)

v10_test_top10 = calculate_top_k_accuracy(
    test_predictions_v10,
    y_test_v10,
    10
)

print("=" * 80)
print("V10 TOP-K ACCURACY")
print("=" * 80)

print(
    f"{'Split':<15}"
    f"{'Top-1':>12}"
    f"{'Top-3':>12}"
    f"{'Top-5':>12}"
    f"{'Top-10':>12}"
)

print("-" * 63)

print(
    f"{'Train':<15}"
    f"{v10_train_top1:>11.4%}"
    f"{v10_train_top3:>11.4%}"
    f"{v10_train_top5:>11.4%}"
    f"{v10_train_top10:>11.4%}"
)

print(
    f"{'Validation':<15}"
    f"{v10_val_top1:>11.4%}"
    f"{v10_val_top3:>11.4%}"
    f"{v10_val_top5:>11.4%}"
    f"{v10_val_top10:>11.4%}"
)

print(
    f"{'Test':<15}"
    f"{v10_test_top1:>11.4%}"
    f"{v10_test_top3:>11.4%}"
    f"{v10_test_top5:>11.4%}"
    f"{v10_test_top10:>11.4%}"
)

print("=" * 80)
```

```python
# ============================================================
# STEP 41.7 — V10 MOST COMMON PREDICTIONS
# ============================================================

from collections import Counter

print("=" * 80)
print("V10 MOST COMMON TEST PREDICTIONS")
print("=" * 80)

prediction_counter_v10 = Counter(
    test_pred_ids_v10
)

print(
    f"{'Rank':<8}"
    f"{'Token ID':<12}"
    f"{'Token':<20}"
    f"{'Count':<12}"
    f"{'Percentage':<12}"
)

print("-" * 64)

for rank, (token_id, count) in enumerate(
    prediction_counter_v10.most_common(20),
    start=1
):

    try:
        token = tokenizer_v8.id_to_token(
            int(token_id)
        )
    except Exception:
        token = "<UNKNOWN>"

    percentage = (
        count / len(test_pred_ids_v10)
    )

    print(
        f"{rank:<8}"
        f"{token_id:<12}"
        f"{token:<20}"
        f"{count:<12}"
        f"{percentage:>10.4%}"
    )

print("=" * 80)
```

```python
# ============================================================
# STEP 41.8 — ACTUAL VS PREDICTED TEST DISTRIBUTION
# ============================================================

print("=" * 80)
print("V10 ACTUAL VS PREDICTED TOKEN DISTRIBUTION")
print("=" * 80)

actual_counter_v10 = Counter(
    y_test_v10.tolist()
)

pred_counter_v10 = Counter(
    test_pred_ids_v10.tolist()
)

all_ids = set(
    actual_counter_v10.keys()
).union(
    pred_counter_v10.keys()
)

actual_total = len(y_test_v10)
pred_total = len(test_pred_ids_v10)

distribution_rows_v10 = []

for token_id in all_ids:

    actual_pct = (
        actual_counter_v10.get(token_id, 0)
        / actual_total
    )

    predicted_pct = (
        pred_counter_v10.get(token_id, 0)
        / pred_total
    )

    difference = (
        predicted_pct - actual_pct
    )

    distribution_rows_v10.append(
        (
            token_id,
            actual_pct,
            predicted_pct,
            difference
        )
    )

# Sort by absolute distribution mismatch

distribution_rows_v10.sort(
    key=lambda x: abs(x[3]),
    reverse=True
)

print(
    f"{'Token':<20}"
    f"{'Actual':>12}"
    f"{'Predicted':>12}"
    f"{'Difference':>14}"
)

print("-" * 60)

for token_id, actual_pct, predicted_pct, difference in \
        distribution_rows_v10[:20]:

    try:
        token = tokenizer_v8.id_to_token(
            int(token_id)
        )
    except Exception:
        token = "<UNKNOWN>"

    print(
        f"{token:<20}"
        f"{actual_pct:>11.4%}"
        f"{predicted_pct:>11.4%}"
        f"{difference:>13.4%}"
    )

print("=" * 80)
```

```python
# ============================================================
# STEP 41.9 — V10 PREDICTION ENTROPY
# ============================================================

def calculate_entropy(predictions):

    clipped = np.clip(
        predictions,
        1e-12,
        1.0
    )

    entropy = -np.sum(
        clipped * np.log(clipped),
        axis=1
    )

    return entropy


v10_train_entropy = calculate_entropy(
    train_predictions_v10
)

v10_val_entropy = calculate_entropy(
    val_predictions_v10
)

v10_test_entropy = calculate_entropy(
    test_predictions_v10
)

max_entropy_v10 = np.log(
    VOCAB_SIZE_V10
)

print("=" * 80)
print("V10 PREDICTION ENTROPY")
print("=" * 80)

print(
    f"Maximum possible entropy : "
    f"{max_entropy_v10:.6f}"
)

print(
    f"\nTrain mean entropy       : "
    f"{v10_train_entropy.mean():.6f}"
)

print(
    f"Val mean entropy         : "
    f"{v10_val_entropy.mean():.6f}"
)

print(
    f"Test mean entropy        : "
    f"{v10_test_entropy.mean():.6f}"
)

print(
    f"\nTrain % of maximum       : "
    f"{100 * v10_train_entropy.mean() / max_entropy_v10:.2f}%"
)

print(
    f"Val % of maximum         : "
    f"{100 * v10_val_entropy.mean() / max_entropy_v10:.2f}%"
)

print(
    f"Test % of maximum        : "
    f"{100 * v10_test_entropy.mean() / max_entropy_v10:.2f}%"
)

print("=" * 80)
```
